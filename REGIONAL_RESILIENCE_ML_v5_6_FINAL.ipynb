```python
# 2. Полный разбор таблиц Росстата → RAW / CLEAN / аналитическая панель
_STAGE2_T0=timing_start("02_full_table_parser")
REGIONS=["Республика Адыгея","Республика Алтай","Республика Башкортостан","Республика Бурятия","Республика Дагестан","Республика Ингушетия","Кабардино-Балкарская Республика","Республика Калмыкия","Карачаево-Черкесская Республика","Республика Карелия","Республика Коми","Республика Марий Эл","Республика Мордовия","Республика Саха (Якутия)","Республика Северная Осетия - Алания","Республика Татарстан","Республика Тыва","Удмуртская Республика","Республика Хакасия","Чеченская Республика","Чувашская Республика","Алтайский край","Забайкальский край","Камчатский край","Краснодарский край","Красноярский край","Пермский край","Приморский край","Ставропольский край","Хабаровский край","Амурская область","Архангельская область","Астраханская область","Белгородская область","Брянская область","Владимирская область","Волгоградская область","Вологодская область","Воронежская область","Ивановская область","Иркутская область","Калининградская область","Калужская область","Кемеровская область","Кировская область","Костромская область","Курганская область","Курская область","Ленинградская область","Липецкая область","Магаданская область","Московская область","Мурманская область","Нижегородская область","Новгородская область","Новосибирская область","Омская область","Оренбургская область","Орловская область","Пензенская область","Псковская область","Ростовская область","Рязанская область","Самарская область","Саратовская область","Сахалинская область","Свердловская область","Смоленская область","Тамбовская область","Тверская область","Томская область","Тульская область","Тюменская область","Ульяновская область","Челябинская область","Ярославская область","г. Москва","г. Санкт-Петербург","Еврейская автономная область","Ненецкий автономный округ","Ханты-Мансийский автономный округ - Югра","Чукотский автономный округ","Ямало-Ненецкий автономный округ","Республика Крым","г. Севастополь"]
ALIASES={"кемеровская область - кузбасс":"Кемеровская область","кемеровская область-кузбасс":"Кемеровская область",
"ханты-мансийский автономный округ":"Ханты-Мансийский автономный округ - Югра","ханты-мансийский автономный округ - югра":"Ханты-Мансийский автономный округ - Югра",
"ханты-мансийский ао - югра":"Ханты-Мансийский автономный округ - Югра","хмао - югра":"Ханты-Мансийский автономный округ - Югра",
"ямало-ненецкий ао":"Ямало-Ненецкий автономный округ","ненецкий ао":"Ненецкий автономный округ","чукотский ао":"Чукотский автономный округ",
"чувашская республика - чувашия":"Чувашская Республика","республика северная осетия-алания":"Республика Северная Осетия - Алания",
"республика северная осетия - алания":"Республика Северная Осетия - Алания",
"москва":"г. Москва","г москва":"г. Москва","город москва":"г. Москва",
"санкт-петербург":"г. Санкт-Петербург","г санкт-петербург":"г. Санкт-Петербург","город санкт-петербург":"г. Санкт-Петербург",
"севастополь":"г. Севастополь","г севастополь":"г. Севастополь"}

from functools import lru_cache

@lru_cache(maxsize=250000)
def norm(s):
    q=str(s).replace("ё","е").replace("Ё","Е").replace("–","-").replace("—","-").replace("\xa0"," ")
    q=re.sub(r"[\*¹²³⁴⁵⁶⁷⁸⁹]+"," ",q)
    q=re.sub(r"\bресп\.?\b","республика",q,flags=re.I)
    q=re.sub(r"\bобл\.?\b","область",q,flags=re.I)
    q=re.sub(r"\bавт\.?\s*окр\.?\b|\bао\b","автономный округ",q,flags=re.I)
    q=re.sub(r"\bг\.(?=\s*[А-Яа-я])","г ",q)
    return re.sub(r"\s+"," ",q).strip(" .;:").lower()

RMAP={norm(r):r for r in REGIONS}; RMAP.update({norm(k):v for k,v in ALIASES.items()})
# Для типичных ячеек Росстата сначала точное сопоставление; дорогой поиск подстроки — только для коротких текстов,
# действительно похожих на наименование территории.
REGION_MARKERS=("республика","область","край","округ","москва","петербург","севастополь")
REGION_PATTERNS=[(k,v,re.compile(r"(?<![а-я])"+re.escape(k)+r"(?![а-я])")) for k,v in sorted(RMAP.items(),key=lambda z:len(z[0]),reverse=True) if len(k)>8]

@lru_cache(maxsize=250000)
def region_of(x):
    if x is None:return None
    s=norm(x)
    if not s or len(s)>180:return None
    v=RMAP.get(s)
    if v:return v
    # Большинство числовых/служебных ячеек отсекаются до regex.
    if not any(m in s for m in REGION_MARKERS):return None
    for _,v,pat in REGION_PATTERNS:
        if pat.search(s):return v
    return None

@lru_cache(maxsize=250000)
def year_of(x):
    m=re.search(r"(?<!\d)(19\d{2}|20[0-2]\d)(?!\d)",str(x)); return int(m.group(1)) if m else None

def number(x):
    if pd.isna(x):return np.nan
    if isinstance(x,(int,float,np.integer,np.floating)):
        try:return float(x)
        except:return np.nan
    st=str(x).replace("\xa0","").replace(" ","").replace(",",".").replace("−","-").strip()
    try:return float(st) if re.fullmatch(r"[-+]?\d+(?:\.\d+)?(?:[eE][-+]?\d+)?",st) else np.nan
    except:return np.nan

def semantic_key(s): return re.sub(r"\b(?:19|20)\d{2}\b|\d+|[^а-яa-z]+"," ",norm(s)).strip()[:220]
def good_text(x):
    s=re.sub(r"\s+"," ",str(x)).strip(); letters=len(re.findall(r"[A-Za-zА-Яа-яЁё]",s)); digits=len(re.findall(r"\d",s)); return bool(s) and not region_of(s) and letters>=4 and letters>=digits*.65

# --- Научно контролируемая семантическая классификация ---
# В V3 использовалось простое совпадение слов. Оно давало ложные отнесения:
# «ВРП на душу населения» -> демография; показатели заболеваемости -> демография и т.п.
# V5 сначала учитывает раздел официального сборника, затем — строгие смысловые признаки.

SOCIAL_SUPPORT_KEYS=[
    "социальн поддерж", "социальн помощ", "субсид", "льгот", "пособ",
    "мер социальной поддержки", "материнск капитал"
]
INCOME_KEYS=[
    "реальн доход", "денежн доход", "среднедушев", "заработн плат",
    "назначенн пенс", "бедност", "прожиточн минимум", "джини",
    "потребительск расход", "жилищно-коммунальн"
]
LABOUR_KEYS=[
    "занятост", "безработ", "рабочей силы", "рабочая сила", "рынок труда",
    "среднегодов численност занят", "потребност в работниках"
]
DEMOGRAPHY_KEYS=[
    "численност населения", "естественн прирост", "миграционн прирост",
    "рождаем", "родивш", "смертност", "умерш", "ожидаемая продолжительность жизни",
    "возрастной состав населения", "демографическ нагруз", "брачност", "разводимост",
    "прибывш", "выбывш", "миграц"
]
ECONOMY_KEYS=[
    "валовой региональный продукт", "индекс физического объема", "промышленн производств",
    "индекс производства", "инвестиц", "объем отгруженных товаров", "сельскохозяйственн производств",
    "строительств", "оборот розничной торговли", "грузооборот", "пассажирооборот",
    "обрабатывающ", "добыч полезных ископаемых"
]
EXCLUDE_HEALTH_EDU_CULTURE=[
    "заболеваем", "больничн", "врач", "медицинск персонал", "здравоохран",
    "образован", "обучающ", "студент", "учител", "культур", "музе", "театр",
    "туризм", "спортивн сооруж", "земельн", "угод", "выброс", "сточн вод"
]

def section_of(title,file="",sheet=""):
    q=" | ".join(map(str,[title,Path(file).stem,sheet]))
    m=re.search(r"раздел\s*(\d+)",norm(q))
    return int(m.group(1)) if m else None

def phrase_hit(q,k):
    return all(tok in q for tok in str(k).split())

def has_any(q,keys):
    return any(phrase_hit(q,k) for k in keys)

def block_of(title,source,file="",sheet=""):
    q=norm(" | ".join(map(str,[title,source,Path(file).stem,sheet])))
    sec=section_of(title,file,sheet)

    # Основной сборник: раздел имеет приоритет над случайными словами в названии.
    if source=="regions_app":
        if sec==1: return "demography"
        if sec==2: return "labour"
        if sec==3:
            if has_any(q,SOCIAL_SUPPORT_KEYS): return "social_support"
            if has_any(q,INCOME_KEYS): return "income"
            return "other"
        # Экономико-производственный блок: только содержательно релевантные показатели.
        if sec in {8,9,10,11,12,13,14,15,16} and has_any(q,ECONOMY_KEYS): return "industry"
        # Финансовые расходы на конкретные меры социальной поддержки допускаются отдельно.
        if sec==19 and has_any(q,SOCIAL_SUPPORT_KEYS): return "social_support"
        return "other"

    # Специализированные источники.
    if source in {"natural_movement","migration"}:
        return "demography" if has_any(q,DEMOGRAPHY_KEYS) else "other"
    if source in {"workforce_compendium","labour_compendium"}:
        if has_any(q,LABOUR_KEYS): return "labour"
        if has_any(q,INCOME_KEYS): return "income"
        return "other"
    if source=="sme_compendium":
        return "industry" if has_any(q,tuple(ECONOMY_KEYS) + ("малое","среднее","предприним","оборот","выручк","инвестиц")) else "other"
    if source.startswith(("vndn_","kouzh_","kdu_")) or source in {"living","social_position"}:
        if has_any(q,SOCIAL_SUPPORT_KEYS): return "social_support"
        if has_any(q,LABOUR_KEYS): return "labour"
        if has_any(q,INCOME_KEYS): return "income"
        return "other"
    if source in {"demography","demographic_yearbook"}:
        if has_any(q,EXCLUDE_HEALTH_EDU_CULTURE): return "other"
        return "demography" if has_any(q,DEMOGRAPHY_KEYS) else "other"

    # Последний безопасный слой: никаких общих слов вроде «население» как самостоятельного признака.
    if has_any(q,EXCLUDE_HEALTH_EDU_CULTURE): return "other"
    if has_any(q,SOCIAL_SUPPORT_KEYS): return "social_support"
    if has_any(q,LABOUR_KEYS): return "labour"
    if has_any(q,DEMOGRAPHY_KEYS): return "demography"
    if has_any(q,INCOME_KEYS): return "income"
    if has_any(q,ECONOMY_KEYS): return "industry"
    return "other"

def semantic_reason(title,source,file="",sheet=""):
    sec=section_of(title,file,sheet)
    return f"source={source}; section={sec if sec is not None else 'NA'}; block={block_of(title,source,file,sheet)}"

def title_from_header(a,start,cols,file,sheet):
    texts=[]
    for r in range(max(0,start-15),start):
        for c in sorted(set([0,1,2]+list(cols))):
            if c<a.shape[1] and good_text(a.iat[r,c]):
                s=re.sub(r"\s+"," ",str(a.iat[r,c])).strip()
                if s not in texts:texts.append(s)
    def score(s):
        q=norm(s); return 5*sum(k in q for k in (SOCIAL_SUPPORT_KEYS+INCOME_KEYS+LABOUR_KEYS+DEMOGRAPHY_KEYS+ECONOMY_KEYS))+min(len(re.findall(r"[а-яa-z]",q)),100)/25
    texts=sorted(texts,key=score,reverse=True); base=" | ".join(texts[:4]); context=" | ".join(x for x in [Path(file).stem,str(sheet)] if good_text(x)); return (base+(" | "+context if context else "")).strip(" |")[:800] or context or Path(file).stem

def make_iid(source,title,file,sheet,slot=""):
    f=semantic_key(Path(file).stem); sh=semantic_key(sheet); t=semantic_key(title); return safe(f"{source}_{f}_{sh}_{t}_{slot}")
def consecutive_groups(xs,max_gap=2):
    xs=sorted(xs)
    if not xs:return []
    out=[]; cur=[xs[0]]
    for x in xs[1:]:
        if x-cur[-1]<=max_gap:cur.append(x)
        else:out.append(cur);cur=[x]
    out.append(cur);return out


# ---------------------------------------------------------------------
# V5. Полный, но быстрый парсер: многолетние и годовые матрицы
# ---------------------------------------------------------------------
# Ключевое исправление V5:
# 1) годовая таблица больше не отбрасывается из-за отсутствия 2+ лет внутри файла;
# 2) один и тот же показатель из ежегодных файлов получает стабильный indicator_id,
#    не зависящий от имени файла и года;
# 3) профилирование охватывает весь разумный размер листа, а не только первые 450 строк;
# 4) сохраняются RAW- и CLEAN-панели и подробный аудит неразобранных листов.

PARSE_ROWS=6000
PARSE_COLS=420
PROFILE_ROWS=5000
PROFILE_COLS=300
HEADER_ROWS=180
REGION_SCAN_COLS=40
MIN_REGION_HITS=5
TARGET_YEARS=set(ANALYSIS_YEARS)

GENERIC_TITLES=(
    "социально-экономические показатели по субъектам российской федерации",
    "социально экономические показатели по субъектам российской федерации",
    "субъекты российской федерации",
    "российская федерация",
    "все население",
)

@lru_cache(maxsize=128)
def file_sha_quick(path_str):
    return sha256(Path(path_str))

def _small(df,rows=PARSE_ROWS,cols=PARSE_COLS):
    if df is None or df.empty:
        return pd.DataFrame()
    return df.iloc[:min(len(df),rows),:min(df.shape[1],cols)]

def _clean_title_piece(x):
    s=re.sub(r"\s+"," ",str(x)).strip()
    q=norm(s)
    if not s or not good_text(s): return ""
    if q in GENERIC_TITLES: return ""
    if q.startswith("итоги выборочного наблюдения доходов населения") and len(q)<90: return ""
    if "федеральный округ" in q and len(q)<100: return ""
    return s

def _title_score(s):
    q=norm(s)
    keys=SOCIAL_SUPPORT_KEYS+INCOME_KEYS+LABOUR_KEYS+DEMOGRAPHY_KEYS+ECONOMY_KEYS
    score=8*sum(phrase_hit(q,k) for k in keys)
    score+=3*bool(re.match(r"^\s*\d+(?:\.\d+)+",s))
    score+=min(len(re.findall(r"[а-яa-z]",q)),140)/28
    if q in GENERIC_TITLES: score-=50
    return score

def title_from_header(a,start,cols,file="",sheet=""):
    """Стабильный содержательный заголовок без имени годового файла."""
    texts=[]
    cols0=sorted(set([0,1,2,3]+[c for c in cols if isinstance(c,(int,np.integer)) and c<a.shape[1]]))
    for r in range(max(0,start-24),min(start+1,a.shape[0])):
        for c in cols0:
            if c>=a.shape[1]: continue
            s=_clean_title_piece(a.iat[r,c])
            if s and s not in texts: texts.append(s)
    texts=sorted(texts,key=_title_score,reverse=True)
    base=" | ".join(texts[:5])
    sh=_clean_title_piece(sheet)
    if not base and sh: base=sh
    return base[:900] if base else "Показатель без распознанного заголовка"

def column_header_path(a,start,c,max_up=18,max_left=10):
    """Иерархический заголовок конкретного столбца с восстановлением объединённых ячеек."""
    vals=[]
    for r in range(max(0,start-max_up),start):
        s=_clean_title_piece(a.iat[r,c]) if c<a.shape[1] else ""
        if not s:
            # В Excel подпись объединённой группы физически хранится в левой ячейке.
            for cc in range(c-1,max(-1,c-max_left)-1,-1):
                z=_clean_title_piece(a.iat[r,cc])
                if z:
                    s=z;break
        if s and s not in vals:vals.append(s)
    # Ближайшие к данным уровни заголовка информативнее дальних.
    return " | ".join(vals[-5:])[:800]

def indicator_signature(title):
    q=norm(title)
    q=re.sub(r"\b(?:19|20)\d{2}\b"," ",q)
    q=re.sub(r"\bтаблица\b|\bтабл\.\b"," ",q)
    q=re.sub(r"\s+"," ",q).strip(" |.;:-")
    # Служебные контексты не должны дробить один показатель по годам.
    for g in GENERIC_TITLES:
        q=q.replace(g," ")
    q=re.sub(r"\s+"," ",q).strip()
    return q[:500]

def make_iid(source,title,file="",sheet="",slot=""):
    sig=indicator_signature(title)
    slot_sig=semantic_key(slot)[:80] if slot else ""
    raw=f"{source}|{sig}|{slot_sig}"
    return f"{safe(source)}_{hashlib.sha1(raw.encode('utf-8')).hexdigest()[:16]}"

def is_technical_title(title):
    q=indicator_signature(title)
    if not q or q=="показатель без распознанного заголовка": return True
    letters=len(re.findall(r"[а-яa-z]",q))
    if letters<8:return True
    if q in GENERIC_TITLES:return True
    if q.startswith("социально-экономические показатели по субъектам") and len(q)<110:return True
    return False

def infer_context_year(df,file="",sheet="",source=None,header_rows=None):
    """Год годовой матрицы: только имя/лист и заголовочная зона, без числового тела таблицы."""
    scores={y:0 for y in ANALYSIS_YEARS}
    fs=f"{Path(file).stem} {sheet}"
    for y in ANALYSIS_YEARS:
        if re.search(rf"(?<!\d){y}(?!\d)",fs):scores[y]+=30
    if df is not None and not df.empty:
        stop=min(len(df),header_rows if header_rows is not None else 35)
        a=df.iloc[:max(0,stop),:min(df.shape[1],24)]
        for x in a.to_numpy().ravel():
            if not isinstance(x,str):continue
            sx=str(x)
            for y in ANALYSIS_YEARS:
                if re.search(rf"(?<!\d){y}(?!\d)",sx):
                    scores[y]+=4 if re.search(r"год|г\.",sx,re.I) else 1
    ranked=sorted(scores.items(),key=lambda z:z[1],reverse=True)
    if ranked and ranked[0][1]>0 and (len(ranked)==1 or ranked[0][1]>ranked[1][1]):
        return ranked[0][0]
    dy=SOURCE_DEFAULT_YEAR.get(source) if source else None
    return int(dy) if dy in ANALYSIS_YEARS else None

def is_row_number_vector(vals):
    x=pd.Series(vals).dropna().to_numpy(float)
    if len(x)<6:return False
    r=np.arange(1,len(x)+1,dtype=float)
    return len(x)==len(r) and np.allclose(x,r,rtol=0,atol=1e-9)

def sheet_profile(df):
    """Один дешёвый проход определяет ориентацию; сканирование шире V4."""
    if df is None or df.empty:return {"mode":"skip","reason":"empty","row_col":None,"header_row":None}
    a=df.iloc[:min(len(df),PROFILE_ROWS),:min(df.shape[1],PROFILE_COLS)]
    nr,nc=a.shape
    row_best=(0,None)
    for c in range(min(nc,REGION_SCAN_COLS)):
        n=sum(region_of(x) is not None for x in a.iloc[:,c].tolist())
        if n>row_best[0]:row_best=(n,c)
    col_best=(0,None)
    for r in range(min(nr,HEADER_ROWS)):
        n=sum(region_of(x) is not None for x in a.iloc[r,:].tolist())
        if n>col_best[0]:col_best=(n,r)
    if row_best[0]>=MIN_REGION_HITS and row_best[0]>=col_best[0]:
        return {"mode":"rows","reason":f"region_rows={row_best[0]}","row_col":row_best[1],"header_row":None}
    if col_best[0]>=MIN_REGION_HITS:
        return {"mode":"columns","reason":f"region_cols={col_best[0]}","row_col":None,"header_row":col_best[1]}
    # Последняя попытка: регион может находиться дальше 40-го столбца.
    for c in range(nc):
        n=sum(region_of(x) is not None for x in a.iloc[:,c].tolist())
        if n>=MIN_REGION_HITS:return {"mode":"long","reason":f"long_region={n}","row_col":c,"header_row":None}
    return {"mode":"skip","reason":f"no_regions(max={max(row_best[0],col_best[0])})","row_col":None,"header_row":None}

def _year_columns_from_header(a,st):
    ycols=[]
    head=a.iloc[max(0,st-28):st,:]
    for c in range(a.shape[1]):
        ys=[year_of(x) for x in head.iloc[:,c].tolist()]
        ys=[y for y in ys if y in TARGET_YEARS]
        if ys:ycols.append((c,ys[-1]))
    return ycols

def parse_region_rows(df,source,file,sheet,rc_hint=None):
    a=_small(df); out=[]
    if a.empty:return out
    if rc_hint is None:
        cand=[]
        for c in range(min(a.shape[1],REGION_SCAN_COLS)):
            hits=[r for r,x in enumerate(a.iloc[:,c].tolist()) if region_of(x)]
            if len(hits)>=MIN_REGION_HITS:cand.append((len(hits),c,hits))
        if not cand:return out
        _,rc,hits=max(cand)
    else:
        rc=int(rc_hint); hits=[r for r,x in enumerate(a.iloc[:,rc].tolist()) if region_of(x)]
        if len(hits)<MIN_REGION_HITS:return out

    context_year=None
    for gi,rows0 in enumerate(consecutive_groups(hits,3)):
        if len(rows0)<MIN_REGION_HITS:continue
        st=min(rows0); context_year=infer_context_year(a,file,sheet,source,header_rows=st); ycols=_year_columns_from_header(a,st)

        # A. Настоящая многолетняя таблица.
        byyear={}
        for c,y in ycols: byyear.setdefault(y,[]).append(c)
        if len(byyear)>=2:
            # Разбиваем на локальные последовательности лет/столбцов.
            runs=[];cur=[];last_c=None;last_y=None
            for c,y in sorted(ycols):
                # Новый блок показателя обычно начинается, когда последовательность лет сбрасывается
                # или между колонками возникает заметный разрыв.
                if cur and (c-last_c>4 or (last_y is not None and y<=last_y)):
                    runs.append(cur);cur=[]
                cur.append((c,y));last_c=c;last_y=y
            if cur:runs.append(cur)
            for ri,run in enumerate(runs):
                if len(set(y for _,y in run))<2:continue
                cols=[c for c,_ in run]
                title=title_from_header(a,st,cols,file,sheet)
                iid=make_iid(source,title,slot=f"multi_{ri}")
                for r in rows0:
                    reg=region_of(a.iat[r,rc])
                    if not reg:continue
                    for c,y in run:
                        v=number(a.iat[r,c])
                        if pd.notna(v):out.append((reg,y,v,iid,title,source,file,sheet))
            continue

        # B. Годовая матрица: один файл/лист = один год, столбцы = разные показатели.
        single_year=(next(iter(byyear)) if len(byyear)==1 else context_year)
        if single_year not in TARGET_YEARS:continue
        # В годовой матрице один заголовок года нередко объединён над несколькими колонками.
        # Поэтому проверяем все колонки и отбираем их уже по фактическому числовому покрытию.
        candidate_cols=[c for c in range(a.shape[1]) if c!=rc]
        for c in candidate_cols:
            vals=[number(a.iat[r,c]) for r in rows0]
            valid=[v for v in vals if pd.notna(v)]
            if len(valid)<max(MIN_REGION_HITS,int(.30*len(rows0))):continue
            if is_row_number_vector(valid):continue
            global_title=title_from_header(a,st,[c],file,sheet)
            specific=column_header_path(a,st,c)
            title=(specific+" | "+global_title if specific and specific not in global_title else (specific or global_title))[:900]
            slot=specific if specific else f"col_{c}"
            iid=make_iid(source,title,slot=slot)
            for r,v in zip(rows0,vals):
                reg=region_of(a.iat[r,rc])
                if reg and pd.notna(v):out.append((reg,single_year,float(v),iid,title,source,file,sheet))
    return out

def parse_region_columns(df,source,file,sheet,hr_hint=None):
    a=_small(df); out=[]
    if a.empty:return out
    if hr_hint is None:
        headers=[]
        for r in range(min(a.shape[0],HEADER_ROWS)):
            cols=[c for c,x in enumerate(a.iloc[r,:].tolist()) if region_of(x)]
            if len(cols)>=MIN_REGION_HITS:headers.append((len(cols),r,cols))
        if not headers:return out
        _,hr,cols=max(headers)
    else:
        hr=int(hr_hint); cols=[c for c,x in enumerate(a.iloc[hr,:].tolist()) if region_of(x)]
        if len(cols)<MIN_REGION_HITS:return out

    left=min(14,a.shape[1]); year_rows=[]
    for r in range(hr+1,a.shape[0]):
        ys=[year_of(x) for x in a.iloc[r,:left].tolist()]
        yy=next((y for y in ys if y in TARGET_YEARS),None)
        if yy:year_rows.append((r,yy))

    # A. Несколько лет внутри таблицы.
    if len(set(y for _,y in year_rows))>=2:
        for gi,grp in enumerate(consecutive_groups([r for r,_ in year_rows],2)):
            pairs=[(r,y) for r,y in year_rows if r in grp]
            if len(set(y for _,y in pairs))<2:continue
            title=title_from_header(a,min(grp),list(range(left)),file,sheet)
            iid=make_iid(source,title,slot=f"rowblock_{gi}")
            for r,y in pairs:
                for c in cols:
                    reg=region_of(a.iat[hr,c]); v=number(a.iat[r,c])
                    if reg and pd.notna(v):out.append((reg,y,v,iid,title,source,file,sheet))
        return out

    # B. Годовая матрица: строки = разные показатели, колонки = регионы.
    single_year=(year_rows[0][1] if year_rows else infer_context_year(a,file,sheet,source,header_rows=hr+1))
    if single_year not in TARGET_YEARS:return out
    start=hr+1
    for r in range(start,a.shape[0]):
        vals=[number(a.iat[r,c]) for c in cols]
        valid=[v for v in vals if pd.notna(v)]
        if len(valid)<max(MIN_REGION_HITS,int(.30*len(cols))):continue
        left_texts=[]
        for c in range(min(left,8)):
            s=_clean_title_piece(a.iat[r,c])
            if s:left_texts.append(s)
        if not left_texts:continue
        title=" | ".join(dict.fromkeys(left_texts))[:900]
        iid=make_iid(source,title,slot=title)
        for c,v in zip(cols,vals):
            reg=region_of(a.iat[hr,c])
            if reg and pd.notna(v):out.append((reg,single_year,float(v),iid,title,source,file,sheet))
    return out

def parse_long(df,source,file,sheet,rc_hint=None):
    """Длинный формат. Использует позиции столбцов — устойчив к повторяющимся заголовкам."""
    if df is None or df.empty:return []
    a=_small(df); ncol=a.shape[1]
    if rc_hint is not None and isinstance(rc_hint,(int,np.integer)) and int(rc_hint)<ncol:
        rc_pos=int(rc_hint)
    else:
        scores=[]
        for j in range(ncol):
            s=a.iloc[:,j]
            scores.append((sum(region_of(x) is not None for x in s.tolist()),j))
        rh,rc_pos=max(scores,default=(0,None))
        if rh<MIN_REGION_HITS:return []
    yc_pos=None;yscore=0
    for j in range(ncol):
        if j==rc_pos:continue
        s=a.iloc[:,j]
        n=sum(year_of(x) in TARGET_YEARS for x in s.tolist())
        if n>yscore:yscore,yc_pos=n,j
    # Годовой файл не обязан иметь отдельный столбец года.
    if yscore<MIN_REGION_HITS:yc_pos=None
    context_year=infer_context_year(a,file,sheet,source,header_rows=min(25,len(a)))
    out=[]
    for j in range(ncol):
        if j in {rc_pos,yc_pos}:continue
        ser=a.iloc[:,j]
        vals=pd.to_numeric(ser,errors="coerce")
        if vals.notna().sum()<MIN_REGION_HITS:continue
        label=str(a.columns[j])
        title=label if good_text(label) else title_from_header(a,0,[j],file,sheet)
        label_dups=sum(str(x)==label for x in list(a.columns))
        slot=(f"{label}|col_{j}" if label_dups>1 else label) if good_text(label) else f"col_{j}"
        iid=make_iid(source,title,slot=slot)
        for i,v in vals[vals.notna()].items():
            reg=region_of(a.iat[i,rc_pos])
            y=year_of(a.iat[i,yc_pos]) if yc_pos is not None else context_year
            if reg and y in TARGET_YEARS:out.append((reg,y,float(v),iid,title,source,file,sheet))
    return out

def read_tables(p):
    """Чтение файла один раз: Excel -> CSV -> HTML. CSV не использует несовместимый low_memory с python engine."""
    suf=p.suffix.lower()
    if suf==".csv":
        raw=p.read_bytes()
        for enc in ("utf-8-sig","utf-8","cp1251","latin1"):
            try:
                txt=raw.decode(enc)
                df=pd.read_csv(io.StringIO(txt),sep=None,engine="python",header=None,on_bad_lines="skip")
                if not df.empty:return {"CSV":df}
            except Exception:pass
        return {}
    if suf in {".html",".htm"}:
        try:
            tabs=pd.read_html(str(p),header=None,displayed_only=False)
            return {f"HTML_{i+1}":d for i,d in enumerate(tabs) if d is not None and not d.empty}
        except Exception:return {}
    if suf in {".xlsx",".xls"}:
        try:return pd.read_excel(p,sheet_name=None,header=None,engine="calamine")
        except Exception:
            eng="openpyxl" if suf==".xlsx" else "xlrd"
            return pd.read_excel(p,sheet_name=None,header=None,engine=eng)
    return {}

panel_cache=PROCESSED/"panel_long.parquet"
raw_cache=PROCESSED/"panel_raw.parquet"
parse_cache=PROCESSED/"table_parse_audit.csv"

if panel_cache.exists() and parse_cache.exists() and not FORCE_PARSE:
    PANEL=pd.read_parquet(panel_cache)
    PANEL_RAW=pd.read_parquet(raw_cache) if raw_cache.exists() else PANEL.copy()
    clean_cache=PROCESSED/"panel_clean.parquet"
    PANEL_CLEAN=pd.read_parquet(clean_cache) if clean_cache.exists() else PANEL.copy()
    TABLE_INDEX=pd.read_csv(parse_cache)
    coll_path=PROCESSED/"indicator_id_collision_audit.csv"
    series_path=PROCESSED/"series_dedup_audit.csv"
    META_COLLISIONS=pd.read_csv(coll_path) if coll_path.exists() else pd.DataFrame()
    SERIES_META=pd.read_csv(series_path) if series_path.exists() else pd.DataFrame()
else:
    rows=[]; audit=[]
    all_files=sorted(
        p for p in EXTRACT.rglob("*")
        if p.is_file() and p.suffix.lower() in TABLE_EXT and valid_table_file(p)
    )
    seen_hash={}; files=[]
    for p in pbar(all_files,desc="2A/5 ХЭШИ И ДУБЛИ",unit="файл",leave=True):
        try:h=file_sha_quick(str(p))
        except Exception:h=f"path:{p}"
        if h in seen_hash:
            audit.append({
                "source":p.relative_to(EXTRACT).parts[0],"file":str(p),"parsed_rows":0,"ok":False,
                "mode":"duplicate_file","sheets":0,"parsed_sheets":0,"seconds":0.0,
                "note":f"duplicate_of={seen_hash[h]}"
            })
            continue
        seen_hash[h]=str(p); files.append(p)

    files=sorted(files,key=lambda p:(0 if "regions_app" in p.parts else 1,str(p)))
    bar=pbar(files,desc="2B/5 РАЗБОР ТАБЛИЦ",unit="файл",leave=True)
    for p in bar:
        t0=time.time(); rel=p.relative_to(EXTRACT); src=rel.parts[0]; n0=len(rows)
        _read_sec=0.0; _parse_sec=0.0
        notes=[]; modes=[]; nsh=0; parsed_sheets=0; sheet_details=[]
        try:
            _read_t0=time.perf_counter(); tabs=read_tables(p); _read_sec=time.perf_counter()-_read_t0
            _parse_t0=time.perf_counter()
            for sh,d in tabs.items():
                bar.set_postfix(src=src,sheet=str(sh)[:18],rows=len(rows)-n0)
                nsh+=1
                prof=sheet_profile(d); mode=prof["mode"]; modes.append(mode)
                if mode=="skip":
                    sheet_details.append(f"{sh}:skip:{prof['reason']}")
                    continue
                got=[]
                try:
                    if mode=="rows":
                        got=parse_region_rows(d,src,str(p),str(sh),prof.get("row_col"))
                    elif mode=="columns":
                        got=parse_region_columns(d,src,str(p),str(sh),prof.get("header_row"))
                    else:
                        got=parse_long(d,src,str(p),str(sh),prof.get("row_col"))
                except Exception as e:
                    notes.append(f"{sh}:{type(e).__name__}:{str(e)[:90]}")

                # Универсальный fallback: до пяти вариантов строки заголовка, но без повторного
                # brute-force сканирования всего файла.
                if not got:
                    try:
                        for h in range(min(5,max(0,len(d)-1))):
                            z=d.iloc[h+1:].copy()
                            z.columns=[str(x) for x in d.iloc[h]]
                            q=parse_long(z.reset_index(drop=True),src,str(p),str(sh))
                            if q:
                                got=q; modes.append("long_fallback"); break
                    except Exception as e:
                        notes.append(f"{sh}:fallback:{type(e).__name__}:{str(e)[:70]}")
                if got:
                    parsed_sheets+=1; rows.extend(got)
                    sheet_details.append(f"{sh}:{mode}:{len(got)}")
                else:
                    sheet_details.append(f"{sh}:{mode}:0")
            _parse_sec=time.perf_counter()-_parse_t0; sec=time.time()-t0
            if sec>SLOW_PARSE_WARN_SEC: notes.append(f"SLOW_FILE>{SLOW_PARSE_WARN_SEC:.0f}s")
            audit.append({
                "source":src,"file":str(p),"parsed_rows":len(rows)-n0,"ok":len(rows)>n0,
                "mode":"/".join(sorted(set(modes))),"sheets":nsh,"parsed_sheets":parsed_sheets,
                "read_seconds":round(_read_sec,3),"parse_seconds":round(_parse_sec,3),"seconds":round(sec,3),
                "note":" | ".join((notes+sheet_details)[-8:])
            })
            bar.set_postfix(src=src,sheets=nsh,rows=len(rows)-n0,sec=f"{sec:.1f}")
        except Exception as e:
            sec=time.time()-t0
            audit.append({
                "source":src,"file":str(p),"parsed_rows":0,"ok":False,"mode":"error",
                "sheets":nsh,"parsed_sheets":parsed_sheets,"read_seconds":round(locals().get("_read_sec",0.0),3),
                "parse_seconds":round(max(0.0,sec-locals().get("_read_sec",0.0)),3),"seconds":round(sec,3),
                "note":f"{type(e).__name__}: {str(e)[:250]}"
            })

    bar.close()
    TABLE_INDEX=pd.DataFrame(audit)
    PANEL_RAW=pd.DataFrame(rows,columns=["region","year","value","indicator_id","indicator","source","file","sheet"])
    if PANEL_RAW.empty:raise RuntimeError("Ни одна таблица не преобразована в региональную панель.")

    PANEL_RAW=PANEL_RAW[
        PANEL_RAW["year"].isin(ANALYSIS_YEARS) & PANEL_RAW["region"].isin(REGIONS)
    ].drop_duplicates(["region","year","indicator_id","value","source","file","sheet"])
    PANEL_RAW["block"]=[
        block_of(i,s0,f,sh)
        for i,s0,f,sh in PANEL_RAW[["indicator","source","file","sheet"]].itertuples(index=False,name=None)
    ]
    PANEL_RAW["technical_title"]=PANEL_RAW["indicator"].map(is_technical_title)

    # --- Канонизация метаданных indicator_id ---
    def _canon_title(vals):
        uniq=list(dict.fromkeys(str(x) for x in vals if str(x).strip()))
        if not uniq:return "Показатель без распознанного заголовка"
        def q(s):
            return (_title_score(s), -len(indicator_signature(s)), -len(s))
        return max(uniq,key=q)
    META_COLLISIONS=PANEL_RAW.groupby("indicator_id").agg(
        title_variants=("indicator","nunique"),block_variants=("block","nunique"),
        source_variants=("source","nunique"),rows=("value","size")
    ).reset_index()
    CANON_TITLE=PANEL_RAW.groupby("indicator_id")["indicator"].agg(_canon_title).to_dict()
    CANON_BLOCK=PANEL_RAW.groupby("indicator_id")["block"].agg(lambda s:s.mode().iat[0] if not s.mode().empty else s.iloc[0]).to_dict()
    CANON_SOURCE=PANEL_RAW.groupby("indicator_id")["source"].agg(lambda s:s.mode().iat[0] if not s.mode().empty else s.iloc[0]).to_dict()
    PANEL_RAW["indicator"]=PANEL_RAW["indicator_id"].map(CANON_TITLE)
    PANEL_RAW["block"]=PANEL_RAW["indicator_id"].map(CANON_BLOCK)
    PANEL_RAW["source"]=PANEL_RAW["indicator_id"].map(CANON_SOURCE)
    PANEL_RAW["technical_title"]=PANEL_RAW["indicator"].map(is_technical_title)

    # CLEAN хранит все содержательные показатели, но уже один показатель = один canonical id/title.
    PANEL_CLEAN=PANEL_RAW[~PANEL_RAW["technical_title"]].copy()
    PANEL_CLEAN=PANEL_CLEAN.groupby(
        ["region","year","indicator_id","indicator","block","source"],as_index=False
    ).agg(value=("value","median"))

    # V5.6: полная очищенная база сохраняется целиком, но модель получает только
    # заранее разрешённые источники. Это не удаляет дополнительную эмпирику.
    PANEL_MODEL_SOURCE=PANEL_CLEAN[PANEL_CLEAN["source"].isin(MODEL_ELIGIBLE_SOURCES)].copy()

    # Точные дубли рядов сохраняются в аудите, но в аналитической панели оставляется один представитель.
    def _series_hash_frame(g):
        q=g[["region","year","value"]].sort_values(["region","year"])
        txt="\n".join(f"{r}|{int(y)}|{float(v):.12g}" for r,y,v in q.itertuples(index=False,name=None))
        return hashlib.sha1(txt.encode("utf-8")).hexdigest()
    SERIES_META=PANEL_MODEL_SOURCE.groupby("indicator_id",sort=False).apply(_series_hash_frame).rename("series_hash").reset_index()
    cc=PANEL_MODEL_SOURCE.groupby("indicator_id").agg(
        indicator=("indicator","first"),block=("block","first"),source=("source","first"),
        regions=("region","nunique"),years=("year","nunique"),obs=("value","size")
    ).reset_index()
    SERIES_META=SERIES_META.merge(cc,on="indicator_id",how="left")
    priority={"regions_app":0,"natural_movement":1,"migration":2,"demographic_yearbook":3,
              "workforce_compendium":4,"labour_compendium":5,"living":6,
              "social_position":7,"sme_compendium":8,"demography":20}
    SERIES_META["source_priority"]=SERIES_META["source"].map(
        lambda s:9 if str(s).startswith("vndn_") else priority.get(s,15)
    )
    SERIES_META["title_quality"]=SERIES_META["indicator"].map(_title_score)
    SERIES_META=SERIES_META.sort_values(
        ["series_hash","years","regions","title_quality","source_priority"],
        ascending=[True,False,False,False,True]
    )
    SERIES_META["duplicate_rank"]=SERIES_META.groupby("series_hash").cumcount()
    SERIES_META["keep_for_model"]=SERIES_META["duplicate_rank"].eq(0)
    keep_ids=set(SERIES_META.loc[SERIES_META["keep_for_model"],"indicator_id"])
    PANEL=PANEL_MODEL_SOURCE[PANEL_MODEL_SOURCE["indicator_id"].isin(keep_ids)].copy()

    PANEL_RAW.to_parquet(raw_cache,index=False)
    PANEL_CLEAN.to_parquet(PROCESSED/"panel_clean.parquet",index=False)
    PANEL.to_parquet(panel_cache,index=False)
    META_COLLISIONS.to_csv(PROCESSED/"indicator_id_collision_audit.csv",index=False)
    SERIES_META.to_csv(PROCESSED/"series_dedup_audit.csv",index=False)
    TABLE_INDEX.to_csv(parse_cache,index=False)

# Комплектность рядов
cov=PANEL.groupby(["indicator_id","indicator","block","source"]).agg(
    regions=("region","nunique"),years=("year","nunique"),obs=("value","size"),
    min_year=("year","min"),max_year=("year","max")
).reset_index()
cov["coverage"]=cov["obs"]/(len(REGIONS)*len(ANALYSIS_YEARS))
cov["full_7y"]=cov["years"].eq(len(ANALYSIS_YEARS))
cov["regions_80plus"]=cov["regions"].ge(80)
cov["research_grade"]=cov["full_7y"] & cov["regions_80plus"]
cov.to_csv(PROCESSED/"indicator_coverage.csv",index=False)

SRC_CLEAN=PANEL_CLEAN.groupby("source").agg(
    rows=("value","size"),regions=("region","nunique"),
    indicators=("indicator_id","nunique"),years=("year","nunique")
).reset_index()
SRC=PANEL.groupby("source").agg(
    rows=("value","size"),regions=("region","nunique"),
    indicators=("indicator_id","nunique"),years=("year","nunique")
).reset_index()

BLK=cov.groupby("block").agg(
    indicators=("indicator_id","nunique"),
    max_regions=("regions","max"),
    max_years=("years","max"),
    research_grade=("research_grade","sum")
).reset_index().sort_values(["research_grade","indicators"],ascending=False)

SEMANTIC_AUDIT=PANEL[["indicator_id","indicator","block","source"]].drop_duplicates().copy()
SEMANTIC_AUDIT["section"]=SEMANTIC_AUDIT["indicator"].map(lambda s: section_of(s))
SEMANTIC_AUDIT["reason"]=SEMANTIC_AUDIT.apply(
    lambda r: semantic_reason(r["indicator"],r["source"]),axis=1
)
SEMANTIC_AUDIT.to_csv(PROCESSED/"semantic_classification_audit.csv",index=False)

TECHNICAL_AUDIT=PANEL_RAW[PANEL_RAW["technical_title"]][
    ["indicator_id","indicator","source","file","sheet"]
].drop_duplicates()
TECHNICAL_AUDIT.to_csv(PROCESSED/"technical_title_audit.csv",index=False)

dup_hashes=set(SERIES_META.loc[SERIES_META.groupby("series_hash")["indicator_id"].transform("size").gt(1),"series_hash"]) if not SERIES_META.empty else set()
DUP_SERIES=SERIES_META[SERIES_META["series_hash"].isin(dup_hashes)].copy() if dup_hashes else pd.DataFrame(columns=SERIES_META.columns if not SERIES_META.empty else [])
DUP_SERIES.to_csv(PROCESSED/"duplicate_series_audit.csv",index=False)

timing_end("02_full_table_parser",_STAGE2_T0,detail=f"raw={len(PANEL_RAW):,}; clean={len(PANEL_CLEAN):,}; model={len(PANEL):,}; indicators={PANEL['indicator_id'].nunique()}")
print("Панель CLEAN:",PANEL_CLEAN.shape,"| аналитическая после точных дублей:",PANEL.shape,"| регионы:",PANEL["region"].nunique(),
      "| показатели:",PANEL["indicator_id"].nunique(),"| годы:",PANEL["year"].nunique())
display(SRC)
display(BLK)
display(cov.sort_values(["research_grade","coverage","regions","years"],ascending=False).head(50))
if "seconds" in TABLE_INDEX.columns and (TABLE_INDEX["seconds"]>0).any():
    timed=TABLE_INDEX.loc[TABLE_INDEX["seconds"]>0,"seconds"]
    print("Скорость: медиана",round(timed.median(),2),"с/файл; p95",round(timed.quantile(.95),2),"с/файл")
    display(TABLE_INDEX.sort_values("seconds",ascending=False).head(20))

pipeline_done("parser","таблицы разобраны")
```

```python
# 2A. АУДИТ КОМПЛЕКТНОСТИ ДАННЫХ — ОБЯЗАТЕЛЬНАЯ КОНТРОЛЬНАЯ ТОЧКА V5.6
# После этой ячейки пользователь оценивает комплектность. Моделирование ещё НЕ запускается.
_STAGE2A_T0=timing_start("02A_data_completeness_audit")

# 1) Комплектность по источникам — используем CLEAN, чтобы точное межисточниковое дублирование не скрывало факт загрузки.
panel_src=PANEL_CLEAN.groupby("source").agg(
    panel_rows=("value","size"),panel_regions=("region","nunique"),
    panel_indicators=("indicator_id","nunique"),panel_years=("year","nunique")
).reset_index()

parse_base=TABLE_INDEX[TABLE_INDEX["mode"].ne("duplicate_file")].copy() if not TABLE_INDEX.empty else pd.DataFrame()
if not parse_base.empty:
    # Классифицируем результат файла: нерегиональная таблица != ошибка парсера.
    def _parse_status(r):
        note=str(r.get("note",""))
        mode=str(r.get("mode",""))
        if bool(r.get("ok",False)):return "parsed"
        if mode=="error" or re.search(r"(XLRDError|ParserError|UnicodeDecodeError|ValueError|AttributeError)",note):return "parser_error"
        if mode in {"rows","columns","long","rows/skip","columns/skip","long/skip"} and re.search(r":(?:rows|columns|long):0",note):
            return "regional_unparsed"
        if "no_regions" in note or mode=="skip":return "nonregional_or_out_of_scope"
        if mode=="duplicate_file":return "duplicate_file"
        return "unclassified_unparsed"
    parse_base["parse_status"]=parse_base.apply(_parse_status,axis=1)
    parse_src=parse_base.groupby("source").agg(
        files_attempted=("file","nunique"),files_with_data=("ok","sum"),
        sheets_seen=("sheets","sum"),sheets_parsed=("parsed_sheets","sum"),
        parsed_rows=("parsed_rows","sum"),median_seconds=("seconds","median"),
        p95_seconds=("seconds",lambda s:s.quantile(.95))
    ).reset_index()
    parse_src["file_parse_rate"]=parse_src["files_with_data"]/parse_src["files_attempted"].replace(0,np.nan)
    parse_src["sheet_parse_rate"]=parse_src["sheets_parsed"]/parse_src["sheets_seen"].replace(0,np.nan)
else:
    parse_src=pd.DataFrame(columns=["source","files_attempted","files_with_data","sheets_seen","sheets_parsed",
                                    "parsed_rows","median_seconds","p95_seconds","file_parse_rate","sheet_parse_rate"])
    parse_base["parse_status"]=[]

SOURCE_COMPLETENESS=SOURCE_AUDIT.merge(parse_src,on="source",how="left").merge(panel_src,on="source",how="left")
for c in ["files_attempted","files_with_data","sheets_seen","sheets_parsed","parsed_rows",
          "panel_rows","panel_regions","panel_indicators","panel_years"]:
    if c in SOURCE_COMPLETENESS:SOURCE_COMPLETENESS[c]=SOURCE_COMPLETENESS[c].fillna(0).astype(int)

# 2) Источник × год + список недостающих регионов.
SOURCE_YEAR=PANEL_CLEAN.groupby(["source","year"]).agg(
    rows=("value","size"),regions=("region","nunique"),indicators=("indicator_id","nunique")
).reset_index()
sr_rows=[]
for (s,y),g in PANEL_CLEAN.groupby(["source","year"]):
    have=set(g["region"])
    miss=[r for r in REGIONS if r not in have]
    sr_rows.append({"source":s,"year":int(y),"regions":len(have),"missing_regions_n":len(miss),
                    "missing_regions":"; ".join(miss)})
SOURCE_YEAR_COVERAGE=pd.DataFrame(sr_rows).sort_values(["source","year"]) if sr_rows else pd.DataFrame()

# 3) Регион × год всей аналитической панели.
REGION_YEAR=PANEL.groupby(["region","year"]).agg(
    indicators=("indicator_id","nunique"),sources=("source","nunique"),observations=("value","size")
).reset_index()
RY_MATRIX=REGION_YEAR.pivot(index="region",columns="year",values="indicators").reindex(index=REGIONS,columns=ANALYSIS_YEARS)

# 4) Комплектность содержательных блоков — cov уже имеет строго одну строку на indicator_id.
BLOCK_COMPLETENESS=cov.groupby("block").agg(
    indicators=("indicator_id","nunique"),research_grade=("research_grade","sum"),
    indicators_80plus=("regions_80plus","sum"),indicators_7y=("full_7y","sum"),
    median_regions=("regions","median"),median_years=("years","median"),
    max_regions=("regions","max"),max_years=("years","max")
).reset_index().sort_values(["research_grade","indicators"],ascending=False)

# 5) Предшоковое и постшоковое покрытие каждого аналитического показателя.
pre_cov=PANEL[PANEL["year"].isin(PRE_YEARS)].groupby("indicator_id").agg(
    pre_regions=("region","nunique"),pre_years=("year","nunique"),pre_obs=("value","size"))
post_cov=PANEL[PANEL["year"].isin(POST_YEARS)].groupby("indicator_id").agg(
    post_regions=("region","nunique"),post_years=("year","nunique"),post_obs=("value","size"))
INDICATOR_COMPLETENESS=cov.merge(pre_cov,left_on="indicator_id",right_index=True,how="left").merge(
    post_cov,left_on="indicator_id",right_index=True,how="left")
for c in ["pre_regions","pre_years","pre_obs","post_regions","post_years","post_obs"]:
    INDICATOR_COMPLETENESS[c]=INDICATOR_COMPLETENESS[c].fillna(0).astype(int)
INDICATOR_COMPLETENESS["usable_trajectory"]=(
    INDICATOR_COMPLETENESS["post_years"].ge(len(POST_YEARS)) &
    INDICATOR_COMPLETENESS["post_regions"].ge(70) &
    INDICATOR_COMPLETENESS["pre_years"].ge(2) &
    INDICATOR_COMPLETENESS["pre_regions"].ge(70))
INDICATOR_COMPLETENESS["usable_predictor"]=(
    INDICATOR_COMPLETENESS["pre_years"].ge(2) &
    INDICATOR_COMPLETENESS["pre_regions"].ge(55))

# 6) Аудит неразобранных файлов: разделяем критические ошибки исследовательского окна,
# исторические/внешние ошибки и штатно пропущенные нерегиональные материалы.
PARSER_FAILURES=parse_base[parse_base["parse_status"].isin(["parser_error","regional_unparsed","unclassified_unparsed"])].copy()
IGNORED_NONREGIONAL=parse_base[parse_base["parse_status"].eq("nonregional_or_out_of_scope")].copy()

def _years_from_row(r):
    txt=" ".join(str(r.get(c,"")) for c in ["file","note","source"])
    return sorted({int(x) for x in re.findall(r"(?<!\d)(20\d{2})(?!\d)",txt)})

if len(PARSER_FAILURES):
    _fy=PARSER_FAILURES.apply(_years_from_row,axis=1)
    _critical=_fy.map(lambda ys: (not ys) or bool(set(ys)&set(ANALYSIS_YEARS)))
    CRITICAL_FAILURES=PARSER_FAILURES[_critical].copy()
    HISTORICAL_FAILURES=PARSER_FAILURES[~_critical].copy()
else:
    CRITICAL_FAILURES=PARSER_FAILURES.copy()
    HISTORICAL_FAILURES=PARSER_FAILURES.copy()
UNPARSED_FILES=CRITICAL_FAILURES.copy()

# 7) Контрольные критерии.
CORE_BLOCKS={"demography","income","labour","industry"}
block_grade=BLOCK_COMPLETENESS.set_index("block")["research_grade"].to_dict()
primary_row=SOURCE_COMPLETENESS[SOURCE_COMPLETENESS["source"].eq("regions_app")]
primary_regions=int(primary_row["panel_regions"].iloc[0]) if len(primary_row) else 0
primary_years=int(primary_row["panel_years"].iloc[0]) if len(primary_row) else 0
primary_rows=int(primary_row["panel_rows"].iloc[0]) if len(primary_row) else 0

GATE=[]
def add_gate(check,required,status,detail):
    GATE.append({"check":check,"required":required,"status":status,"detail":detail})

add_gate("official_primary_source",True,
         "ok" if (len(primary_row) and bool(primary_row["ok"].iloc[0])) else "fail",
         "Основной официальный архив Росстата")
add_gate("panel_regions",True,"ok" if PANEL["region"].nunique()>=80 else "fail",
         f"{PANEL['region'].nunique()} регионов")
add_gate("panel_years_2017_2023",True,
         "ok" if set(ANALYSIS_YEARS).issubset(set(PANEL["year"].unique())) else "fail",
         "годы="+",".join(map(str,sorted(PANEL["year"].unique()))))
add_gate("primary_longitudinal_coverage",True,
         "ok" if primary_regions>=80 and primary_years==len(ANALYSIS_YEARS) else "fail",
         f"regions_app: rows={primary_rows}; regions={primary_regions}; years={primary_years}")
for b in sorted(CORE_BLOCKS):
    n=int(block_grade.get(b,0))
    add_gate(f"core_block_{b}",True,"ok" if n>=2 else ("warning" if n>=1 else "fail"),
             f"полных уникальных рядов 80+ регионов × 7 лет: {n}")

# ВНДН оценивается по всем доступным годовым выпускам, а не только 2023/2024.
vndn_rows=SOURCE_COMPLETENESS[SOURCE_COMPLETENESS["source"].astype(str).str.startswith("vndn_")]
vndn_loaded=int((vndn_rows["panel_rows"]>0).sum()) if len(vndn_rows) else 0
add_gate("vndn_historical_coverage",False,
         "ok" if vndn_loaded>=4 else ("warning" if vndn_loaded>=2 else "warning"),
         f"выпусков ВНДН с региональными данными: {vndn_loaded} из 7 (2018–2024)")

for s in ["natural_movement","migration","living","social_position","demographic_yearbook","workforce_compendium","labour_compendium","sme_compendium","demography"]:
    rr=SOURCE_COMPLETENESS[SOURCE_COMPLETENESS["source"].eq(s)]
    if rr.empty:add_gate(f"source_{s}",False,"warning","источник отсутствует в аудите")
    else:
        pr=int(rr["panel_rows"].iloc[0]);rg=int(rr["panel_regions"].iloc[0]);yr=int(rr["panel_years"].iloc[0])
        st="ok" if pr>0 and rg>=70 else "warning"
        add_gate(f"source_{s}",False,st,f"panel_rows={pr}; regions={rg}; years={yr}")

critical_failures=len(CRITICAL_FAILURES)
add_gate("parser_failures_target_window",False,"ok" if critical_failures==0 else "warning",
         f"критических ошибок 2017–2023/неопределённого года: {critical_failures}; исторических/внешних: {len(HISTORICAL_FAILURES)}; нерегиональных пропущено штатно: {len(IGNORED_NONREGIONAL)}")
add_gate("technical_titles_removed",False,"ok" if len(TECHNICAL_AUDIT)==0 else "warning",
         f"технических заголовков исключено из CLEAN: {len(TECHNICAL_AUDIT)}")
add_gate("exact_duplicate_series_removed_from_model",False,"ok" if len(DUP_SERIES)==0 else "warning",
         f"строк аудита дублей: {len(DUP_SERIES)}; в PANEL оставлен один представитель каждой точной серии")
collision_bad=int(((META_COLLISIONS.get("block_variants",pd.Series(dtype=int))>1) |
                   (META_COLLISIONS.get("source_variants",pd.Series(dtype=int))>1)).sum()) if not META_COLLISIONS.empty else 0
add_gate("indicator_id_collisions",False,"ok" if collision_bad==0 else "warning",
         f"indicator_id с конфликтом блока/источника: {collision_bad}")

DATA_GATE=pd.DataFrame(GATE)
DATA_RECOMMENDED_READY=not ((DATA_GATE["required"]) & (DATA_GATE["status"].eq("fail"))).any()

# 8) Экспорт контрольного комплекта ДО моделирования.
DATA_AUDIT_XLSX=RESULTS/"CHAPTER3_DATA_AUDIT_V5_6.xlsx"
with pd.ExcelWriter(DATA_AUDIT_XLSX,engine="xlsxwriter") as w:
    RUN_TIMINGS=timing_frame()
    sheets={
        "RUN_TIMINGS":RUN_TIMINGS,"DATA_GATE":DATA_GATE,"SOURCE_AUDIT":SOURCE_AUDIT,"SOURCE_COMPLETENESS":SOURCE_COMPLETENESS,
        "DISCOVERY_AUDIT":DISCOVERY_AUDIT,"DOWNLOAD_INDEX":DOWNLOAD_INDEX,"PARSE_AUDIT":TABLE_INDEX,
        "SOURCE_CATALOG":SOURCE_CATALOG,
        "PARSER_FAILURES":PARSER_FAILURES,"CRITICAL_FAILURES":CRITICAL_FAILURES,
        "HISTORICAL_FAILURES":HISTORICAL_FAILURES,"IGNORED_NONREGIONAL":IGNORED_NONREGIONAL,
        "SOURCE_YEAR":SOURCE_YEAR,"SOURCE_YEAR_COVERAGE":SOURCE_YEAR_COVERAGE,
        "REGION_YEAR":REGION_YEAR,"REGION_YEAR_MATRIX":RY_MATRIX.reset_index(),
        "BLOCK_COMPLETENESS":BLOCK_COMPLETENESS,"INDICATOR_COMPLETENESS":INDICATOR_COMPLETENESS,
        "SEMANTIC_AUDIT":SEMANTIC_AUDIT,"TECHNICAL_TITLES":TECHNICAL_AUDIT,
        "INDICATOR_ID_AUDIT":META_COLLISIONS,"DUPLICATE_SERIES":DUP_SERIES,
    }
    for name,df in sheets.items():df.to_excel(w,sheet_name=name[:31],index=False)

# V5.6: тяжёлый DATA_STAGE ZIP намеренно не создаётся.
timing_end("02A_data_completeness_audit",_STAGE2A_T0,detail=f"gate_ready={DATA_RECOMMENDED_READY}; critical_failures={len(CRITICAL_FAILURES)}")
RUN_TIMINGS=timing_frame(); RUN_TIMINGS.to_csv(PROCESSED/"run_timings.csv",index=False)
try:
    with pd.ExcelWriter(DATA_AUDIT_XLSX,engine="openpyxl",mode="a",if_sheet_exists="replace") as _tw:
        RUN_TIMINGS.to_excel(_tw,sheet_name="RUN_TIMINGS",index=False)
except Exception as _e:
    print("Предупреждение: не удалось обновить лист RUN_TIMINGS:",_e)
print("\n=== КОНТРОЛЬНАЯ ТОЧКА ДАННЫХ V5.6 ===")
display(DATA_GATE)
print("\nКомплектность источников:");display(SOURCE_COMPLETENESS)
print("\nКомплектность содержательных блоков:");display(BLOCK_COMPLETENESS)
print("\nКритические ошибки парсера 2017–2023 / неопределённого года:");display(CRITICAL_FAILURES.head(50))
print("\nРЕКОМЕНДАЦИЯ АВТОМАТИЧЕСКОГО GATE:",
      "МОЖНО РАССМАТРИВАТЬ ПРОДОЛЖЕНИЕ" if DATA_RECOMMENDED_READY else "НЕ ПРОДОЛЖАТЬ ДО ИСПРАВЛЕНИЯ ОБЯЗАТЕЛЬНЫХ ПУНКТОВ")
print("Окончательное решение принимает исследователь.")
print("Контрольный Excel:",DATA_AUDIT_XLSX)
print("\nТайминги этапов данных:"); display(timing_frame())

pipeline_done("data_gate",f"DATA_GATE={DATA_RECOMMENDED_READY}; регионы={PANEL.region.nunique()}")
```

```python
# 3. Постшоковые траектории устойчивости -> типология только при доказанной воспроизводимости
if not globals().get("CONTINUE_RESEARCH",False):
    print("Блок пропущен: CONTINUE_RESEARCH=False. Это штатная остановка после аудита данных.")
else:
    _STAGE3_T0=timing_start("03_postshock_trajectory_space")
    OUTCOME_BLOCKS=["demography","income","labour","industry"]
    PREDICTOR_BLOCKS=["demography","income","labour","industry","social_support"]
    TARGET_BLOCKS=OUTCOME_BLOCKS  # совместимость функций блока 3
    ry=PANEL[PANEL.year.isin(ANALYSIS_YEARS)].groupby(["indicator_id","region"]).year.nunique().reset_index(name="ny")
    usable=ry[ry.ny>=5].groupby("indicator_id").region.nunique().rename("usable_regions")

    # V5.6 FIX: полный coverage сохраняется для объясняющей модели.
    # Раньше COV сразу урезался до outcome-блоков, поэтому social_support
    # физически исчезал еще ДО whitelist блока 4.
    COV_ALL=cov.copy()
    COV_ALL["usable_regions"]=COV_ALL.indicator_id.map(usable).fillna(0).astype(int)
    COV_ALL["sem"]=COV_ALL.indicator.astype(str).map(semantic_key)

    # Для траекторий используем только outcome-блоки.
    COV_TRAJECTORY=COV_ALL[COV_ALL.block.isin(TARGET_BLOCKS)].copy()
    COV=COV_TRAJECTORY  # совместимость оставшейся логики блока 3

    TRAJECTORY_PRIORITY={
     "demography":["естественн прирост","миграционн прирост","ожидаем продолжительност жизни","суммарн коэффициент рождаем","общ коэффициент смертност","возрастн состав","демографическ нагруз"],
     "income":["реальн денежн доход","реальн начисленн заработн плат","реальн размер назначенн пенс","границ бедност","прожиточн минимум","среднедушев денежн доход","потребительск расход"],
     "labour":["уровен безработиц","уровен занятост","уровен участи состав рабоч сил","среднегодов численност занят","численност безработн"],
     "industry":["индекс физическ объем валов региональн продукт","индекс производств","индекс физическ объем инвестиц","промышленн производств","инвестиц основн капитал","оборот розничн торговл"],
     "social_support":["числ семей получавш субсид","среднемесячн размер субсид","численност граждан пользующ социальн поддерж","мер социальн поддерж","социальн поддерж"]
    }
    TRAJECTORY_CODES={
     "demography":[r"(^|\|)\s*1\.14\.",r"(^|\|)\s*1\.20\.",r"(^|\|)\s*1\.16\.1\.",r"(^|\|)\s*1\.15\."],
     "income":[r"(^|\|)\s*3\.1\.1\.",r"(^|\|)\s*3\.1\.2\.",r"(^|\|)\s*3\.1\.3\.",r"(^|\|)\s*3\.12\."],
     "labour":[r"(^|\|)\s*2\.10\.1\.",r"(^|\|)\s*2\.5\.",r"(^|\|)\s*2\.2\.",r"(^|\|)\s*2\.3\."],
     "industry":[r"(^|\|)\s*8\.3\.",r"(^|\|)\s*10\.3\.",r"(^|\|)\s*12\.3\.",r"(^|\|)\s*12\.5\."],
     "social_support":[r"(^|\|)\s*3\.28\.1\.",r"(^|\|)\s*3\.29\.1\.",r"(^|\|)\s*19\.3\."]
    }
    NOMINAL_BAD=["в текущих ценах","миллионов рублей","тысяч рублей","рублей на душу","общая сумма начисленных субсидий","объем средств"]
    RELATIVE_GOOD=["процент","коэффициент","индекс","реальн","на 1000","на 10 000","удельный вес","уровень"]

    def trajectory_score(row):
        raw=str(row.indicator); q=norm(raw); pri=TRAJECTORY_PRIORITY.get(str(row.block),[])
        code_hits=[400-25*rank for rank,pat in enumerate(TRAJECTORY_CODES.get(str(row.block),[])) if re.search(pat,raw,re.I)]
        sem_hits=[100-8*rank for rank,k in enumerate(pri) if phrase_hit(q,k)]
        score=max(code_hits or [0])+max(sem_hits or [0])
        if has_any(q,RELATIVE_GOOD): score+=18
        if has_any(q,NOMINAL_BAD) and not has_any(q,["реальн","индекс","процент"]): score-=45
        score+=min(float(row.usable_regions),85)/20+min(float(row.years),7)
        return score
    COV["trajectory_score"]=COV.apply(trajectory_score,axis=1)

    def choose_indicators(z,per_block=4):
        for mr,my in [(65,6),(55,6),(45,5),(35,5)]:
            q=z[(z.usable_regions>=mr)&(z.years>=my)].sort_values(["block","trajectory_score","usable_regions","coverage"],ascending=[True,False,False,False])
            out=[]
            for bl in TARGET_BLOCKS:
                seen=set()
                for _,r in q[q.block==bl].iterrows():
                    if r.trajectory_score<15: continue
                    sk=r["sem"]
                    if sk in seen: continue
                    seen.add(sk); out.append(r)
                    if sum(str(x["block"])==bl for x in out)>=per_block: break
            s=pd.DataFrame(out,columns=z.columns).drop_duplicates("indicator_id")
            core={"demography","income","labour","industry"}
            if len(s)>=12 and core.issubset(set(s.block)) and (s.groupby("block").size().reindex(list(core)).fillna(0)>=3).all():
                return s.reset_index(drop=True),mr,my
        return pd.DataFrame(columns=z.columns),None,None

    SELECTED,SEL_MIN_REG,SEL_MIN_YEARS=choose_indicators(COV)
    if SELECTED.empty: raise RuntimeError("Недостаточное покрытие для построения пространства траекторий.")
    SELECTED=SELECTED.drop_duplicates("indicator_id").reset_index(drop=True)

    # V5.6: направление outcome задается ТОЛЬКО по содержательному показателю.
    # Общие заголовки разделов («Занятость и безработица») не используются,
    # чтобы уровень занятости/участия в рабочей силе не инвертировался ошибочно.
    DIRECTION_RULES=[
        ("demography", r"(^|\|)\s*1\.14\.", +1, "естественный прирост"),
        ("demography", r"(^|\|)\s*1\.20\.", +1, "миграционный прирост"),
        ("demography", r"(^|\|)\s*1\.16\.1\.", +1, "ожидаемая продолжительность жизни"),
        ("demography", r"(^|\|)\s*1\.15\.", +1, "суммарная рождаемость"),
        ("demography", r"коэффициент.*смертност|смертност.*на 1000", -1, "смертность"),
        ("income", r"(^|\|)\s*3\.1\.1\.", +1, "реальные доходы"),
        ("income", r"(^|\|)\s*3\.1\.2\.", +1, "реальная заработная плата"),
        ("income", r"(^|\|)\s*3\.1\.3\.", +1, "реальные пенсии"),
        ("income", r"(^|\|)\s*3\.12\.", -1, "бедность"),
        ("labour", r"(^|\|)\s*2\.10\.1\.", -1, "уровень безработицы"),
        ("labour", r"(^|\|)\s*2\.5\.", +1, "уровень занятости"),
        ("labour", r"(^|\|)\s*2\.2\.", +1, "участие в рабочей силе"),
        ("labour", r"(^|\|)\s*2\.3\.", +1, "среднегодовая численность занятых"),
        ("industry", r"(^|\|)\s*8\.3\.", +1, "физический объем ВРП"),
        ("industry", r"(^|\|)\s*10\.3\.", +1, "физический объем инвестиций"),
        ("industry", r"(^|\|)\s*12\.3\.", +1, "промышленное производство"),
        ("industry", r"(^|\|)\s*12\.5\.", +1, "обрабатывающие производства"),
    ]

    def trajectory_direction(block,title):
        raw=str(title)
        for bl,pat,direction,reason in DIRECTION_RULES:
            if str(block)==bl and re.search(pat,raw,re.I):
                return int(direction), reason
        # Без молчаливого угадывания: неизвестный outcome должен быть рассмотрен вручную.
        raise ValueError(f"Не задано направление outcome: block={block}; indicator={title}")

    _dir=[trajectory_direction(b,t) for b,t in zip(SELECTED["block"],SELECTED["indicator"])]
    SELECTED["direction"]=[x[0] for x in _dir]
    SELECTED["direction_reason"]=[x[1] for x in _dir]
    DIRECTION_AUDIT=SELECTED[["indicator_id","indicator","block","direction","direction_reason"]].copy()
    META=SELECTED.set_index("indicator_id")[["indicator","block","direction"]].to_dict("index")
    regs=sorted(set(REGIONS)&set(PANEL.region.unique()))

    def build_deviation_scheme(scale_mode):
        dev_long=[]; metric_rows=[]; used=[]
        for iid in pbar(SELECTED["indicator_id"].tolist(),desc=f"3A/6 НОРМИРОВАНИЕ {scale_mode}",unit="показ.",leave=True):
            q=PANEL[(PANEL.indicator_id==iid)&PANEL.year.isin(ANALYSIS_YEARS)].pivot_table(index="region",columns="year",values="value",aggfunc="median").reindex(index=regs,columns=ANALYSIS_YEARS)
            good=(q.notna().sum(axis=1)>=5)&(q[PRE_YEARS].notna().sum(axis=1)>=2)&(q[POST_YEARS].notna().sum(axis=1)>=2)
            q=q.loc[good].copy()
            if len(q)<30: continue
            q[PRE_YEARS]=q[PRE_YEARS].apply(lambda r:r.interpolate(limit_direction="both"),axis=1)
            q[POST_YEARS]=q[POST_YEARS].apply(lambda r:r.interpolate(limit_direction="both"),axis=1)
            pre=q[PRE_YEARS]; center=pre.median(axis=1)
            if scale_mode=="region_robust":
                mad=pre.sub(center,axis=0).abs().median(axis=1)*1.4826
                std=pre.std(axis=1)
                scale=pd.concat([mad,std],axis=1).max(axis=1)
                floor=np.maximum(np.abs(center)*.03,1e-9)
                scale=pd.Series(np.maximum(scale.fillna(0).to_numpy(),floor.to_numpy()),index=q.index)
                dev=q.sub(center,axis=0).div(scale,axis=0)
            else:
                resid=pre.sub(center,axis=0).to_numpy(float).ravel(); resid=resid[np.isfinite(resid)]
                medr=float(np.nanmedian(resid)) if len(resid) else 0.0
                gmad=float(np.nanmedian(np.abs(resid-medr))*1.4826) if len(resid) else 0.0
                gstd=float(np.nanstd(resid)) if len(resid) else 0.0
                level=float(np.nanmedian(np.abs(center.to_numpy(float)))) if len(center) else 0.0
                sc=max(gmad,gstd,level*.01,1e-9)
                dev=q.sub(center,axis=0)/sc
            dev=dev.clip(-8,8)
            indicator=str(META[iid]["indicator"]); block=str(META[iid]["block"])
            direction=int(META[iid].get("direction",1))
            dev=dev*direction
            used.append(iid)
            for reg in dev.index:
                post=dev.loc[reg,POST_YEARS].to_numpy(dtype=float); finite=np.isfinite(post)
                if finite.sum()<2: continue
                inside=np.where(np.abs(post)<=.75)[0]
                metric_rows.append({"scale_mode":scale_mode,"region":str(reg),"indicator_id":str(iid),"indicator":indicator,"block":block,
                                    "shock_2020":float(post[0]),"max_abs_deviation":float(np.nanmax(np.abs(post))),
                                    "recovery_lag_years":float(inside[0]) if len(inside) else np.nan,
                                    "late_level":float(np.nanmean(post[-2:])),
                                    "post_slope":float(np.polyfit(np.arange(len(post))[finite],post[finite],1)[0]) if finite.sum()>1 else np.nan,
                                    "post_volatility":float(np.nanstd(post))})
                dev_long.extend({"scale_mode":scale_mode,"region":str(reg),"year":int(y0),"indicator_id":str(iid),"block":block,"deviation":float(v)} for y0,v in zip(POST_YEARS,post))
        return pd.DataFrame(dev_long),pd.DataFrame(metric_rows),used

    DEV_POOLED,METRICS_POOLED,used_pooled=build_deviation_scheme("pooled_robust")
    DEV_REGION,METRICS_REGION,used_region=build_deviation_scheme("region_robust")
    used=set(used_pooled)&set(used_region)
    SELECTED=SELECTED[SELECTED.indicator_id.isin(used)].drop_duplicates("indicator_id").reset_index(drop=True)
    if len(SELECTED)<12: raise RuntimeError(f"После нормирования осталось только {len(SELECTED)} показателей.")

    def block_year_matrix(dev,chosen_ids=None,drop_blocks=None):
        z=dev.copy()
        if chosen_ids is not None: z=z[z.indicator_id.isin(set(map(str,chosen_ids)))]
        if drop_blocks: z=z[~z.block.isin(set(drop_blocks))]
        a=z.groupby(["region","block","year"],as_index=False).deviation.median()
        w=a.assign(col=a.block+"__"+a.year.astype(str)).pivot(index="region",columns="col",values="deviation")
        expected=[f"{b}__{y0}" for b in TARGET_BLOCKS if not drop_blocks or b not in drop_blocks for y0 in POST_YEARS]
        w=w.reindex(columns=expected)
        w=w.loc[w.notna().mean(axis=1)>=.75]
        return w.apply(lambda c:c.fillna(c.median()),axis=0).fillna(0)

    def indicator_year_matrix(dev):
        z=dev.assign(col=dev.indicator_id.astype(str)+"__"+dev.year.astype(str))
        w=z.pivot_table(index="region",columns="col",values="deviation",aggfunc="median")
        w=w.loc[w.notna().mean(axis=1)>=.70]
        return w.apply(lambda c:c.fillna(c.median()),axis=0).fillna(0)

    def block_metric_matrix(metrics):
        cols=["shock_2020","late_level","post_slope","post_volatility"]
        a=metrics.groupby(["region","block"],as_index=False)[cols].median()
        parts=[]
        for st in cols:
            w=a.pivot(index="region",columns="block",values=st); w.columns=[f"{b}__{st}" for b in w.columns]; parts.append(w)
        z=pd.concat(parts,axis=1); z=z.loc[z.notna().mean(axis=1)>=.75]
        return z.apply(lambda c:c.fillna(c.median()),axis=0).fillna(0)

    TRAJ_PRIMARY=block_year_matrix(DEV_POOLED)
    TRAJ_REGION=block_year_matrix(DEV_REGION)
    TRAJ_INDICATOR=indicator_year_matrix(DEV_POOLED)
    TRAJ_METRICS=block_metric_matrix(METRICS_POOLED)
    if len(TRAJ_PRIMARY)<40: raise RuntimeError(f"В основном пространстве осталось только {len(TRAJ_PRIMARY)} регионов.")

    def embed_matrix(M,seed=SEED,variance=.90):
        X=StandardScaler().fit_transform(M)
        ncomp=max(2,min(20,X.shape[0]-1,X.shape[1]))
        p=PCA(n_components=ncomp,random_state=seed); E0=p.fit_transform(X)
        n90=max(2,min(E0.shape[1],int(np.searchsorted(np.cumsum(p.explained_variance_ratio_),variance)+1)))
        return E0[:,:n90],E0,p

    def cluster_fixed(M,k,seed=SEED):
        E,_,_=embed_matrix(M,seed); lab=KMeans(n_clusters=k,n_init=100,random_state=seed).fit_predict(E)
        return lab,E

    def evaluate_rep(name,M):
        rows=[]
        E,_,_=embed_matrix(M)
        for k in range(2,min(5,len(M)-1)):
            lab=KMeans(n_clusters=k,n_init=100,random_state=SEED).fit_predict(E); sz=pd.Series(lab).value_counts()
            if sz.min()<8: continue
            rows.append({"representation":name,"k":k,"silhouette":silhouette_score(E,lab),"min_cluster":int(sz.min()),"max_cluster":int(sz.max()),"regions":len(M),"features":M.shape[1]})
        return rows

    REPRESENTATION_ROWS=[]
    for name,M in [("block_year_pooled",TRAJ_PRIMARY),("block_year_region",TRAJ_REGION),("block_metrics_pooled",TRAJ_METRICS),("indicator_year_pooled",TRAJ_INDICATOR)]:
        REPRESENTATION_ROWS.extend(evaluate_rep(name,M))
    REPRESENTATION_AUDIT=pd.DataFrame(REPRESENTATION_ROWS)
    prim=REPRESENTATION_AUDIT[REPRESENTATION_AUDIT.representation=="block_year_pooled"].sort_values("silhouette",ascending=False)
    if prim.empty: raise RuntimeError("Основное блочное представление не дало допустимого разбиения.")
    BEST_K=int(prim.iloc[0].k)
    PRIMARY_SIL=float(prim.iloc[0].silhouette)
    base_labels,E=cluster_fixed(TRAJ_PRIMARY,BEST_K,SEED)
    primary_regions=TRAJ_PRIMARY.index.astype(str).tolist()

    # Альтернативная кластеризация и чувствительность к нормированию.
    try:
        alt=AgglomerativeClustering(n_clusters=BEST_K,linkage="ward").fit_predict(E); ALT_ARI=adjusted_rand_score(base_labels,alt)
    except Exception: ALT_ARI=np.nan; alt=None
    common=TRAJ_PRIMARY.index.intersection(TRAJ_REGION.index)
    norm_ari=np.nan; norm_labels=None
    if len(common)>=40:
        lr,_=cluster_fixed(TRAJ_REGION.loc[common],BEST_K,SEED+77)
        lp,_=cluster_fixed(TRAJ_PRIMARY.loc[common],BEST_K,SEED)
        norm_ari=adjusted_rand_score(lp,lr); norm_labels=lr

    stab=[]; partitions=[]; continuous_stability=[]
    partitions.append(("primary",base_labels.copy()))
    for seed in pbar(range(50),desc="3B/6 УСТОЙЧИВОСТЬ ПО SEED",unit="повт.",leave=True):
        z,_=cluster_fixed(TRAJ_PRIMARY,BEST_K,1000+seed); stab.append({"mode":"seed","rep":seed,"ARI":adjusted_rand_score(base_labels,z)}); partitions.append((f"seed_{seed}",z))

    rng=np.random.default_rng(SEED)
    ids_by_block={b:SELECTED[SELECTED.block==b].indicator_id.astype(str).tolist() for b in TARGET_BLOCKS}
    for r in pbar(range(BOOT_N),desc="3C/6 СТРАТИФИЦИРОВАННЫЕ ПОДВЫБОРКИ",unit="повт.",leave=True):
        take=[]
        for bl,ids in ids_by_block.items():
            if not ids: continue
            n=max(1,min(len(ids),int(np.ceil(.75*len(ids)))))
            take.extend(rng.choice(np.array(ids,dtype=object),size=n,replace=False).tolist())
        M=block_year_matrix(DEV_POOLED,take).reindex(primary_regions).dropna(how="all")
        if len(M)!=len(primary_regions): continue
        z,Esub=cluster_fixed(M,BEST_K,SEED+r); ari=adjusted_rand_score(base_labels,z)
        stab.append({"mode":"indicator_stratified","rep":r,"ARI":ari}); partitions.append((f"indsub_{r}",z))
        try:
            rho=float(spearmanr(pdist(E[:,:2]),pdist(Esub[:,:2])).statistic)
        except Exception:
            rho=np.nan
        continuous_stability.append({"mode":"indicator_stratified","rep":r,"distance_rho":rho})

    loo=[]
    for iid in pbar(SELECTED.indicator_id.astype(str).tolist(),desc="3D/6 LEAVE-ONE-INDICATOR",unit="показ.",leave=True):
        take=[x for x in SELECTED.indicator_id.astype(str).tolist() if x!=iid]
        M=block_year_matrix(DEV_POOLED,take).reindex(primary_regions)
        if M.isna().all(axis=1).any(): continue
        z,Esub=cluster_fixed(M.fillna(M.median()).fillna(0),BEST_K,SEED+31); ari=adjusted_rand_score(base_labels,z)
        loo.append({"mode":"leave_one_indicator","rep":iid,"ARI":ari}); partitions.append((f"loo_{iid}",z))
        try:
            rho=float(spearmanr(pdist(E[:,:2]),pdist(Esub[:,:2])).statistic)
        except Exception:
            rho=np.nan
        continuous_stability.append({"mode":"leave_one_indicator","rep":iid,"distance_rho":rho})

    lob=[]
    for bl in pbar(TARGET_BLOCKS,desc="3E/6 LEAVE-ONE-BLOCK",unit="блок",leave=True):
        M=block_year_matrix(DEV_POOLED,drop_blocks=[bl]).reindex(primary_regions)
        if M.isna().all(axis=1).any(): continue
        z,Esub=cluster_fixed(M.fillna(M.median()).fillna(0),BEST_K,SEED+47); ari=adjusted_rand_score(base_labels,z)
        lob.append({"mode":"leave_one_block","rep":bl,"ARI":ari})
        try:
            rho=float(spearmanr(pdist(E[:,:2]),pdist(Esub[:,:2])).statistic)
        except Exception:
            rho=np.nan
        continuous_stability.append({"mode":"leave_one_block","rep":bl,"distance_rho":rho})
    CLUSTER_STABILITY=pd.concat([pd.DataFrame(stab),pd.DataFrame(loo),pd.DataFrame(lob)],ignore_index=True)

    # Consensus/co-association matrix по возмущениям, не зависящая от нумерации меток.
    n=len(primary_regions); co=np.zeros((n,n),float)
    for _,lab in partitions: co+=(lab[:,None]==lab[None,:]).astype(float)
    co/=max(len(partitions),1); dist=1-co; np.fill_diagonal(dist,0)
    try:
        cons=AgglomerativeClustering(n_clusters=BEST_K,metric="precomputed",linkage="average").fit_predict(dist)
    except TypeError:
        cons=AgglomerativeClustering(n_clusters=BEST_K,affinity="precomputed",linkage="average").fit_predict(dist)
    CONSENSUS_ARI=float(adjusted_rand_score(base_labels,cons)); CONSENSUS_SIL=float(silhouette_score(E,cons))
    same=(cons[:,None]==cons[None,:]); tri=np.triu(np.ones_like(same,dtype=bool),1)
    within=float(np.nanmean(co[tri & same])) if np.any(tri & same) else np.nan
    between=float(np.nanmean(co[tri & ~same])) if np.any(tri & ~same) else np.nan
    CONSENSUS_GAP=within-between if pd.notna(within) and pd.notna(between) else np.nan
    CONSENSUS_MATRIX=pd.DataFrame(co,index=primary_regions,columns=primary_regions).reset_index(names="region")

    indicator_med=float(CLUSTER_STABILITY.loc[CLUSTER_STABILITY["mode"].eq("indicator_stratified"),"ARI"].median())
    seed_med=float(CLUSTER_STABILITY.loc[CLUSTER_STABILITY["mode"].eq("seed"),"ARI"].median())
    loo_med=float(CLUSTER_STABILITY.loc[CLUSTER_STABILITY["mode"].eq("leave_one_indicator"),"ARI"].median())
    lob_med=float(CLUSTER_STABILITY.loc[CLUSTER_STABILITY["mode"].eq("leave_one_block"),"ARI"].median())

    # Устойчивость непрерывной геометрии при другой нормировке.
    norm_distance_rho=np.nan
    if len(common)>=40:
        try:
            _,Epool=cluster_fixed(TRAJ_PRIMARY.loc[common],BEST_K,SEED)
            _,Ereg=cluster_fixed(TRAJ_REGION.loc[common],BEST_K,SEED+77)
            norm_distance_rho=float(spearmanr(pdist(Epool[:,:2]),pdist(Ereg[:,:2])).statistic)
        except Exception:
            norm_distance_rho=np.nan

    CONTINUUM_STABILITY=pd.DataFrame(continuous_stability)
    ind_geom=float(CONTINUUM_STABILITY.loc[CONTINUUM_STABILITY["mode"].eq("indicator_stratified"),"distance_rho"].median()) if len(CONTINUUM_STABILITY) else np.nan
    loo_geom=float(CONTINUUM_STABILITY.loc[CONTINUUM_STABILITY["mode"].eq("leave_one_indicator"),"distance_rho"].median()) if len(CONTINUUM_STABILITY) else np.nan
    lob_geom=float(CONTINUUM_STABILITY.loc[CONTINUUM_STABILITY["mode"].eq("leave_one_block"),"distance_rho"].median()) if len(CONTINUUM_STABILITY) else np.nan

    TYPOLOGY_ACCEPTED=bool(
        PRIMARY_SIL>=TYPOLOGY_MIN_SILHOUETTE and
        indicator_med>=TYPOLOGY_MIN_INDICATOR_ARI and
        pd.notna(norm_ari) and norm_ari>=TYPOLOGY_MIN_NORMALIZATION_ARI and
        CONSENSUS_ARI>=TYPOLOGY_MIN_CONSENSUS_ARI and
        pd.notna(CONSENSUS_GAP) and CONSENSUS_GAP>=TYPOLOGY_MIN_CONSENSUS_GAP
    )

    CONTINUUM_ACCEPTED=bool(
        pd.notna(norm_distance_rho) and norm_distance_rho>=CONTINUUM_MIN_NORM_DISTANCE_RHO and
        pd.notna(ind_geom) and ind_geom>=CONTINUUM_MIN_INDICATOR_DISTANCE_RHO and
        pd.notna(loo_geom) and loo_geom>=CONTINUUM_MIN_LOO_DISTANCE_RHO and
        pd.notna(lob_geom) and lob_geom>=CONTINUUM_MIN_LOB_DISTANCE_RHO
    )

    # V5.6: кластеризация остаётся диагностикой. Даже если конкретный прогон
    # формально прошёл кластерный gate, основной научный результат строится
    # как непрерывное/многомерное пространство, поскольку это устойчивее к
    # составу показателей и не вводит искусственный индекс.
    if CONTINUUM_ACCEPTED:
        ANALYSIS_MODE="multidimensional_continuous_space"
    else:
        ANALYSIS_MODE="multidimensional_trajectory_profiles"

    final_labels=cons if TYPOLOGY_ACCEPTED else base_labels
    CLUSTERS=pd.DataFrame({"region":primary_regions,"cluster":final_labels.astype(int)+1})
    CLUSTERS["type"]="Тип "+CLUSTERS.cluster.astype(str)
    CLUSTERS["typology_accepted"]=TYPOLOGY_ACCEPTED

    # Непрерывное пространство и нагрузки PCA.
    _,E0,PCA_MODEL=embed_matrix(TRAJ_PRIMARY)
    pca_scores=E0[:,:2].copy()

    # Знак PC1 делаем интерпретируемым: положительное направление связано
    # с более благоприятным поздним уровнем outcome-блоков (после direction harmonisation).
    late_cols=[f"{b}__2023" for b in OUTCOME_BLOCKS if f"{b}__2023" in TRAJ_PRIMARY.columns]
    if late_cols:
        anchor=TRAJ_PRIMARY[late_cols].mean(axis=1).to_numpy(float)
        try:
            rr=float(spearmanr(pca_scores[:,0],anchor).statistic)
        except Exception:
            rr=np.nan
        if pd.notna(rr) and rr<0:
            pca_scores[:,0]*=-1
            PCA_MODEL.components_[0,:]*=-1

    PCA_COORD=pd.DataFrame(pca_scores,index=TRAJ_PRIMARY.index.astype(str),columns=["PC1","PC2"]).reset_index(names="region").merge(CLUSTERS,on="region")
    PCA_VARIANCE=pd.DataFrame({
        "component":np.arange(1,len(PCA_MODEL.explained_variance_ratio_)+1),
        "explained_variance":PCA_MODEL.explained_variance_ratio_,
        "cumulative":np.cumsum(PCA_MODEL.explained_variance_ratio_)
    })
    load_rows=[]
    for ci in range(min(2,PCA_MODEL.components_.shape[0])):
        for feature,val in zip(TRAJ_PRIMARY.columns,PCA_MODEL.components_[ci]):
            load_rows.append({"component":f"PC{ci+1}","feature":feature,"loading":float(val)})
    PCA_LOADINGS=pd.DataFrame(load_rows)
    CONTINUOUS_TARGETS=PCA_COORD[["region","PC1","PC2"]].copy()

    # Блочные непрерывные оси: fallback, если общей воспроизводимой геометрии нет.
    block_target=pd.DataFrame({"region":TRAJ_PRIMARY.index.astype(str)})
    block_load=[]
    block_var=[]
    for bl in OUTCOME_BLOCKS:
        cols=[f"{bl}__{y0}" for y0 in POST_YEARS if f"{bl}__{y0}" in TRAJ_PRIMARY.columns]
        if len(cols)<3:
            continue
        M=TRAJ_PRIMARY[cols].copy()
        Z=StandardScaler().fit_transform(M)
        pp=PCA(n_components=1,random_state=SEED)
        score=pp.fit_transform(Z).ravel()
        # Положительный знак = более благоприятный поздний outcome.
        anchor=M[cols[-1]].to_numpy(float)
        try:
            rr=float(spearmanr(score,anchor).statistic)
        except Exception:
            rr=np.nan
        if pd.notna(rr) and rr<0:
            score*=-1
            pp.components_[0,:]*=-1
        block_target[f"{bl}_axis"]=score
        block_var.append({"block":bl,"explained_variance":float(pp.explained_variance_ratio_[0])})
        for f,v in zip(cols,pp.components_[0]):
            block_load.append({"block":bl,"feature":f,"loading":float(v)})
    BLOCK_TARGETS=block_target
    BLOCK_PCA_VARIANCE=pd.DataFrame(block_var)
    BLOCK_PCA_LOADINGS=pd.DataFrame(block_load)

    METRICS=METRICS_POOLED[METRICS_POOLED.region.isin(primary_regions)].copy(); DEV_LONG=DEV_POOLED[DEV_POOLED.region.isin(primary_regions)].copy()
    PM=METRICS.merge(CLUSTERS,on="region",how="inner")
    PROFILES=PM.groupby(["cluster","type","block"],as_index=False).agg(n_regions=("region","nunique"),shock_2020=("shock_2020","median"),max_abs_deviation=("max_abs_deviation","median"),recovery_lag_years=("recovery_lag_years","median"),late_level=("late_level","median"),post_slope=("post_slope","median"),post_volatility=("post_volatility","median"))
    PLOT_TRAJECTORIES=DEV_LONG.merge(CLUSTERS,on="region").groupby(["block","type","year"],as_index=False).deviation.median()
    TYPOLOGY_DIAGNOSTICS=pd.DataFrame([
        {"metric":"analysis_mode","value":ANALYSIS_MODE},
        {"metric":"primary_silhouette","value":PRIMARY_SIL},
        {"metric":"seed_median_ARI","value":seed_med},
        {"metric":"indicator_stratified_median_ARI","value":indicator_med},
        {"metric":"leave_one_indicator_median_ARI","value":loo_med},
        {"metric":"leave_one_block_median_ARI","value":lob_med},
        {"metric":"normalization_ARI","value":norm_ari},
        {"metric":"alternative_hierarchical_ARI","value":ALT_ARI},
        {"metric":"consensus_ARI","value":CONSENSUS_ARI},
        {"metric":"consensus_silhouette","value":CONSENSUS_SIL},
        {"metric":"consensus_within","value":within},
        {"metric":"consensus_between","value":between},
        {"metric":"consensus_gap","value":CONSENSUS_GAP},
        {"metric":"typology_accepted","value":TYPOLOGY_ACCEPTED},
        {"metric":"continuum_norm_distance_rho","value":norm_distance_rho},
        {"metric":"continuum_indicator_median_distance_rho","value":ind_geom},
        {"metric":"continuum_loo_indicator_median_distance_rho","value":loo_geom},
        {"metric":"continuum_leave_block_median_distance_rho","value":lob_geom},
        {"metric":"continuum_accepted","value":CONTINUUM_ACCEPTED},
    ])

    TRAJ_PRIMARY.to_parquet(PROCESSED/"trajectory_block_year_primary.parquet")
    METRICS.to_parquet(PROCESSED/"trajectory_metrics.parquet",index=False)
    CLUSTERS.to_csv(PROCESSED/"regional_types_exploratory.csv",index=False)
    CONTINUOUS_TARGETS.to_csv(PROCESSED/"continuous_trajectory_targets.csv",index=False)
    BLOCK_TARGETS.to_csv(PROCESSED/"block_trajectory_targets.csv",index=False)
    CONTINUUM_STABILITY.to_csv(PROCESSED/"continuous_geometry_stability.csv",index=False)
    SELECTED.to_csv(PROCESSED/"selected_trajectory_indicators.csv",index=False)

    plt.figure(figsize=(8.5,6.2))
    if TYPOLOGY_ACCEPTED:
        for t,g in PCA_COORD.groupby("type"): plt.scatter(g.PC1,g.PC2,label=t,s=48,alpha=.82)
        plt.legend(frameon=False)
    else:
        plt.scatter(PCA_COORD.PC1,PCA_COORD.PC2,s=48,alpha=.82)
    plt.xlabel("Первая компонента траекторий"); plt.ylabel("Вторая компонента траекторий"); plt.tight_layout()
    for ext in ("png","svg"): plt.savefig(FIG/f"01_trajectory_space.{ext}",dpi=600 if ext=="png" else None,bbox_inches="tight")
    plt.close()

    RU_BLOCK={"demography":"Демографическая динамика","income":"Доходы и уровень жизни","labour":"Рынок труда","industry":"Экономическая и отраслевая динамика","social_support":"Социальная поддержка"}
    if TYPOLOGY_ACCEPTED:
        for bl,g in pbar(list(PLOT_TRAJECTORIES.groupby("block")),desc="3F/6 ГРАФИКИ ТРАЕКТОРИЙ",unit="блок",leave=True):
            plt.figure(figsize=(9,5.4))
            for t,h in g.groupby("type"): plt.plot(h.year,h.deviation,marker="o",label=t)
            plt.axvline(SHOCK_YEAR,color="black",lw=1,ls="--"); plt.axhline(0,color="black",lw=.7); plt.title(RU_BLOCK.get(bl,bl)); plt.xlabel("Год"); plt.ylabel("Медианное стандартизированное отклонение"); plt.legend(frameon=False); plt.tight_layout()
            for ext in ("png","svg"): plt.savefig(FIG/f"trajectory_{bl}.{ext}",dpi=600 if ext=="png" else None,bbox_inches="tight")
            plt.close()

    timing_end("03_postshock_trajectory_space",_STAGE3_T0,detail=f"mode={ANALYSIS_MODE}; regions={len(primary_regions)}; indicators={len(SELECTED)}; k={BEST_K}; silhouette={PRIMARY_SIL:.3f}; indicator_ARI={indicator_med:.3f}")
    print(f"V5.6: mode={ANALYSIS_MODE} | regions={len(primary_regions)} | indicators={len(SELECTED)} | k={BEST_K} | silhouette={PRIMARY_SIL:.3f} | indicator ARI={indicator_med:.3f} | continuum norm rho={norm_distance_rho:.3f} | continuum indicator rho={ind_geom:.3f}")
    display(REPRESENTATION_AUDIT.sort_values(["representation","silhouette"],ascending=[True,False])); display(TYPOLOGY_DIAGNOSTICS); display(SELECTED[["block","indicator","usable_regions","years","trajectory_score"]])

pipeline_done("trajectories",f"траектории: mode={ANALYSIS_MODE}; continuum={CONTINUUM_ACCEPTED}")
```

```python
# 4. Предшоковые факторы -> семантически строгая антициркулярная модель и проверки устойчивости
if not globals().get("CONTINUE_RESEARCH",False):
    print("Блок пропущен: CONTINUE_RESEARCH=False.")
else:
    _STAGE4_T0=timing_start("04_explanatory_models_and_robustness")
    STRICT_ANTICIRCULAR=True
    # Только research-grade ряды из ПОЛНОГО coverage.
    # COV_TRAJECTORY содержит только outcome, поэтому для факторов используем COV_ALL.
    _PRED_COV=COV_ALL if "COV_ALL" in globals() else cov.copy()
    if "usable_regions" not in _PRED_COV.columns:
        _PRED_COV=_PRED_COV.copy()
        _PRED_COV["usable_regions"]=_PRED_COV.indicator_id.map(usable).fillna(0).astype(int)
    PRED=_PRED_COV[
        (_PRED_COV["research_grade"].eq(True) if "research_grade" in _PRED_COV.columns else pd.Series(False,index=_PRED_COV.index))
        &(_PRED_COV.usable_regions>=35)
        &(_PRED_COV.years>=3)
    ].copy()
    preobs=PANEL[PANEL.year.isin(PRE_YEARS)].groupby("indicator_id").agg(pre_regions=("region","nunique"),pre_years=("year","nunique"),pre_obs=("value","size"))
    PRED=PRED.merge(preobs,left_on="indicator_id",right_index=True,how="left"); PRED=PRED[(PRED.pre_regions>=35)&(PRED.pre_years>=2)].copy()
    selected_sem=set(SELECTED.indicator.astype(str).map(semantic_key)); PRED["sem"]=PRED.indicator.astype(str).map(semantic_key)
    if STRICT_ANTICIRCULAR:
        PRED=PRED[(~PRED.indicator_id.isin(set(SELECTED.indicator_id)))&(~PRED["sem"].isin(selected_sem))].copy()

    PREDICTOR_PRIORITY={
      "demography":["удельный вес городского населения","население старше трудоспособного возраста","население в трудоспособном возрасте","коэффициенты демографической нагрузки","смертность населения в трудоспособном возрасте","соотношение мужчин и женщин"],
      "income":["среднедушевые денежные доходы","медианный среднедушевой денежный доход","среднемесячная номинальная начисленная заработная плата","потребительские расходы в среднем на душу","численность занятых, приходящихся на одного пенсионера","удельный вес расходов домашних хозяйств на оплату жилищно-коммунальных услуг"],
      "labour":["уровень зарегистрированной безработицы","потребность в работниках","нагрузка незанятого населения","изменение среднегодовой численности занятых","безработицы населения в возрасте 15-29 лет","иностранных граждан"],
      "industry":["валовой региональный продукт на душу населения","инвестиции в основной капитал на душу населения","удельный вес индивидуальных предпринимателей","ввод в действие жилых домов","структура оборота розничной торговли","индексы производства продукции сельского хозяйства"],
      "social_support":["среднемесячный размер социальной поддержки","среднемесячный размер субсидий на семью","объем средств, предусмотренных на предоставление гражданам социальной поддержки","число семей, получавших субсидии","расходы консолидированных бюджетов субъектов российской федерации на реализацию мер социальной поддержки","численность граждан, пользующихся социальной поддержкой","общая сумма начисленных субсидий"]
    }
    SOCIAL_FORBIDDEN=["лук","лимон","сапог","ботин","костюм","говядин","свинин","куры","молоко","масло","хлеб","товары, шт","покупательная способность"]
    SOCIAL_SUPPORT_CODES=[
        r"(^|\\|)\\s*3\\.28\\.[123]\\.",
        r"(^|\\|)\\s*3\\.29\\.[123]\\.",
        r"(^|\\|)\\s*19\\.3\\."
    ]

    def predictor_score(r):
        q=norm(r.indicator); pats=PREDICTOR_PRIORITY.get(str(r.block),[]); hits=[]
        for rank,p in enumerate(pats):
            if norm(p) in q:
                hits.append(150-15*rank)

        # Для социальной поддержки используем также устойчивые к формулировкам
        # коды таблиц основного сборника Росстата: 3.28.*, 3.29.*, 19.3.
        if str(r.block)=="social_support":
            for rank,pat in enumerate(SOCIAL_SUPPORT_CODES):
                if re.search(pat,str(r.indicator),re.I):
                    hits.append(180-10*rank)

        if not hits:
            return -999
        sc=max(hits)+min(float(r.pre_regions),85)/10+min(float(r.pre_years),3)*3
        if str(r.block)=="social_support" and has_any(q,SOCIAL_FORBIDDEN): return -999
        # Предпочитаем относительные/душевые структурные характеристики; абсолютный масштаб допустим только при отсутствии лучшего ряда.
        if has_any(q,["в процентах","на 1000","на душу","удельный вес","коэффициент","уровень"]): sc+=12
        if has_any(q,["миллионов рублей","тысяч человек","на конец года; человек"]) and not has_any(q,["на 1000","на душу","в процентах"]): sc-=12
        return sc

    # Явная проверка: блок социальной поддержки должен присутствовать
    # уже до scoring. Это отделяет проблему данных от проблемы whitelist.
    SOCIAL_SUPPORT_CANDIDATES=PRED[PRED.block.eq("social_support")].copy()
    if len(SOCIAL_SUPPORT_CANDIDATES)==0:
        raise RuntimeError(
            "В полном research-grade coverage отсутствуют кандидаты social_support. "
            "Это уже проблема данных/классификации, а не whitelist."
        )

    PRED["predictor_score"]=PRED.apply(predictor_score,axis=1)
    SOCIAL_SUPPORT_CANDIDATES=PRED[PRED.block.eq("social_support")][
        ["indicator_id","indicator","source","predictor_score","pre_regions","pre_years","research_grade"]
    ].sort_values(["predictor_score","pre_regions"],ascending=[False,False]).reset_index(drop=True)

    print("\nКандидаты блока social_support перед окончательным отбором:")
    display(SOCIAL_SUPPORT_CANDIDATES.head(20))

    if int((SOCIAL_SUPPORT_CANDIDATES.predictor_score>0).sum())<3:
        raise RuntimeError(
            "Данные social_support найдены, но whitelist/code-selector пропустил менее 3 показателей. "
            "Смотрите SOCIAL_SUPPORT_CANDIDATES."
        )

    pred_ids=[]; pred_meta={}; selected_rows=[]
    for bl in PREDICTOR_BLOCKS:
        g=PRED[(PRED.block==bl)&(PRED.predictor_score>0)].sort_values(["predictor_score","pre_regions","coverage"],ascending=[False,False,False])
        seen=set()
        for _,r in g.iterrows():
            sk=semantic_key(r.indicator)
            if sk in seen: continue
            seen.add(sk); pred_ids.append(r.indicator_id); pred_meta[r.indicator_id]=(bl,r.indicator); selected_rows.append(r)
            limit=3 if bl=="social_support" else 4
            if sum(pred_meta[x][0]==bl for x in pred_ids)>=limit: break
    PREDICTOR_SELECTION=pd.DataFrame(selected_rows).drop_duplicates("indicator_id") if selected_rows else pd.DataFrame()
    required_pred_counts={"demography":3,"income":3,"labour":3,"industry":3,"social_support":3}
    counts=pd.Series([pred_meta[x][0] for x in pred_ids],dtype="object").value_counts()
    bad={b:int(counts.get(b,0)) for b,nmin in required_pred_counts.items() if int(counts.get(b,0))<nmin}
    PREDICTOR_BLOCK_AUDIT=pd.DataFrame([
        {"block":b,"selected":int(counts.get(b,0)),"required_min":nmin,
         "status":"ok" if int(counts.get(b,0))>=nmin else "fail"}
        for b,nmin in required_pred_counts.items()
    ])
    if bad:
        display(PRED[PRED.block.isin(required_pred_counts)].sort_values(
            ["block","predictor_score"],ascending=[True,False]
        )[["block","indicator","predictor_score","pre_regions","pre_years","research_grade"]].head(120))
        raise RuntimeError(f"После содержательного whitelist недостаточно факторов по блокам: {bad}")

    pred_code={iid:f"p{j:03d}" for j,iid in enumerate(pred_ids,1)}
    STAT_RU={"mean":"уровень","slope":"тренд"}
    def short_title(s,n=72):
        x=re.sub(r"\s+"," ",str(s)).strip(); parts=[p.strip() for p in x.split("|") if p.strip()]
        parts=[p for p in parts if "социально-экономические показатели" not in norm(p) and "федеральный округ" not in norm(p) and "раздел " not in norm(p)]
        x=parts[0] if parts else x
        return x if len(x)<=n else x[:n-1]+"…"

    def needs_log(title):
        q=norm(title)
        return has_any(q,["рублей","тысяч человек","миллионов рублей","человек"]) and not has_any(q,["процент","на 1000","на 10 000","на душу","коэффициент","индекс","удельный вес","уровень"])

    # V5.6: одинаково тестируем общие компоненты и четыре содержательных оси.
    MODEL_TARGETS=(CONTINUOUS_TARGETS[["region","PC1","PC2"]]
                   .merge(BLOCK_TARGETS,on="region",how="inner"))

    def build_predictors(years,show_progress=False,label="ПРЕДИКТОРЫ"):
        rows=[]; _regs=MODEL_TARGETS.region.tolist(); _iter=pbar(_regs,desc=label,unit="рег.",leave=True) if show_progress else _regs
        for reg in _iter:
            rr={"region":reg}
            for iid in pred_ids:
                bl,title=pred_meta[iid]
                vals=PANEL[(PANEL.region==reg)&(PANEL.indicator_id==iid)&PANEL.year.isin(years)].groupby("year").value.median().reindex(years).to_numpy(float)
                if needs_log(title):
                    vals=np.where(vals>=0,np.log1p(vals),np.nan)
                code=pred_code[iid]; finite=np.isfinite(vals)
                rr[f"{bl}__{code}__mean"]=np.nanmean(vals) if finite.any() else np.nan
                rr[f"{bl}__{code}__slope"]=np.polyfit(np.arange(len(vals))[finite],vals[finite],1)[0] if finite.sum()>=2 else np.nan
            rows.append(rr)
        return pd.DataFrame(rows).merge(MODEL_TARGETS,on="region",how="inner").merge(CLUSTERS[["region","cluster","type"]],on="region",how="left")

    FEATURE_META=[]
    for iid,(bl,title) in pred_meta.items():
        for st in ["mean","slope"]:
            FEATURE_META.append({"feature":f"{bl}__{pred_code[iid]}__{st}","indicator_id":iid,"block":bl,"stat":st,"indicator":title,"transform":"log1p" if needs_log(title) else "raw","label":short_title(title)+" — "+STAT_RU[st]})
    FEATURE_META=pd.DataFrame(FEATURE_META)

    XDF=build_predictors(PRE_YEARS,show_progress=True,label="4A/7 ПРЕДШОКОВЫЕ ПРИЗНАКИ")
    feature_cols=[c for c in XDF if "__" in c and XDF[c].notna().mean()>=.55 and XDF[c].nunique(dropna=True)>1]
    # Корреляционный отсев проводится ВНУТРИ блоков. Межблочную корреляцию не используем
    # как повод удалить целый теоретически заданный механизм (особенно social_support).
    keep=[]
    for bl in PREDICTOR_BLOCKS:
        cols=[c for c in feature_cols if c.startswith(bl+"__")]
        if not cols:
            continue
        completeness={c:float(XDF[c].notna().mean()) for c in cols}
        ordered=sorted(cols,key=lambda c:(-completeness[c],c))
        corr=XDF[ordered].corr(method="spearman",min_periods=20).abs()
        kept_bl=[]
        for c in ordered:
            if not any(pd.notna(corr.loc[c,k]) and corr.loc[c,k]>.85 for k in kept_bl):
                kept_bl.append(c)
        keep.extend(kept_bl)
    feature_cols=keep
    if len(feature_cols)<10:
        raise RuntimeError("После контроля пропусков/внутриблочных корреляций осталось слишком мало предикторов.")
    X=XDF[feature_cols]
    blocks={bl:[c for c in feature_cols if c.startswith(bl+"__")] for bl in PREDICTOR_BLOCKS}
    missing_model_blocks=[b for b in PREDICTOR_BLOCKS if len(blocks.get(b,[]))==0]
    if missing_model_blocks:
        raise RuntimeError(f"После подготовки признаков исчезли обязательные блоки: {missing_model_blocks}")

    def cb_cls(seed=SEED,iters=600): return CatBoostClassifier(iterations=iters,depth=5,learning_rate=.04,loss_function="MultiClass",auto_class_weights="Balanced",random_seed=seed,verbose=False,allow_writing_files=False)
    def cb_reg(seed=SEED,iters=600): return CatBoostRegressor(iterations=iters,depth=5,learning_rate=.04,loss_function="RMSE",random_seed=seed,verbose=False,allow_writing_files=False)

    # ------------------------------------------------------------------
    # V5.6. Этап 1: одинаковая первичная валидация всех непрерывных целей.
    # ------------------------------------------------------------------
    TARGETS=["PC1","PC2"]+[f"{b}_axis" for b in OUTCOME_BLOCKS if f"{b}_axis" in XDF.columns]
    TARGET_LABELS={
        "PC1":"Общая компонента PC1",
        "PC2":"Общая компонента PC2",
        "demography_axis":"Демографическая траектория",
        "income_axis":"Траектория доходов",
        "labour_axis":"Траектория рынка труда",
        "industry_axis":"Экономическая траектория",
    }
    CV_ROWS=[]; LOYO_ROWS=[]; BOOT_ROWS=[]; NEG_ROWS=[]
    PERM_ROWS=[]; SHAP_ROWS=[]; LINEAR_ROWS=[]; ABL_ROWS=[]
    PREDICTIONS=XDF[["region"]].copy()

    def _fit_predict_pair(Xtr,Xte,ytr,seed):
        im=SimpleImputer(strategy="median",keep_empty_features=True)
        a=im.fit_transform(Xtr); b=im.transform(Xte)
        sc=StandardScaler(); ar=sc.fit_transform(a); br=sc.transform(b)
        rg=Ridge(alpha=1.0).fit(ar,ytr)
        cb=cb_reg(seed,iters=450); cb.fit(a,ytr)
        return rg.predict(br),cb.predict(b)

    for ti,target in enumerate(TARGETS,1):
        y=XDF[target].to_numpy(float)
        splits=list(RepeatedKFold(
            n_splits=5,n_repeats=REPEATED_CV_REPEATS,random_state=SEED
        ).split(X,y))
        pred_sum=np.zeros(len(y),float); pred_n=np.zeros(len(y),int)
        for f,(tr,te) in enumerate(pbar(
            splits,desc=f"6A CV {ti}/{len(TARGETS)} {target}",unit="fold",leave=False
        )):
            pl,pc=_fit_predict_pair(X.iloc[tr],X.iloc[te],y[tr],SEED+100*ti+f)
            pred_sum[te]+=pc; pred_n[te]+=1
            for name,p in [("Ridge",pl),("CatBoost",pc)]:
                CV_ROWS.append({
                    "target":target,"fold":f,"model":name,
                    "r2":r2_score(y[te],p),"mae":mean_absolute_error(y[te],p),
                    "n_test":len(te)
                })
        PREDICTIONS[target]=y
        PREDICTIONS[target+"_catboost_pred"]=pred_sum/np.maximum(pred_n,1)

        # LOYO: каждый предшоковый год исключается из построения факторов.
        for held in PRE_YEARS:
            years=[yy for yy in PRE_YEARS if yy!=held]
            D=build_predictors(years)
            cols=[c for c in feature_cols if c in D.columns and D[c].notna().mean()>=.50 and D[c].nunique(dropna=True)>1]
            yy=D[target].to_numpy(float)
            vals=[]
            for tr,te in KFold(5,shuffle=True,random_state=SEED+held+10*ti).split(D[cols],yy):
                im=SimpleImputer(strategy="median",keep_empty_features=True)
                a=im.fit_transform(D.iloc[tr][cols]); b=im.transform(D.iloc[te][cols])
                m=cb_reg(SEED+held+10*ti,iters=350); m.fit(a,yy[tr])
                vals.append(r2_score(yy[te],m.predict(b)))
            LOYO_ROWS.append({
                "target":target,"held_year":held,"r2":float(np.mean(vals)),"features":len(cols)
            })

    CV_METRICS=pd.DataFrame(CV_ROWS)
    TEMPORAL=pd.DataFrame(LOYO_ROWS)

    # Первичный screening: только положительные и лучше линейного baseline цели
    # проходят к дорогим robustness-проверкам.
    screen_rows=[]
    for target in TARGETS:
        cbv=CV_METRICS[(CV_METRICS.target==target)&(CV_METRICS.model=="CatBoost")].r2.astype(float)
        rgv=CV_METRICS[(CV_METRICS.target==target)&(CV_METRICS.model=="Ridge")].r2.astype(float)
        ly=TEMPORAL[TEMPORAL.target.eq(target)].r2.astype(float)
        row={
            "target":target,
            "label":TARGET_LABELS.get(target,target),
            "cv_r2_mean":float(cbv.mean()),
            "cv_r2_median":float(cbv.median()),
            "cv_positive_share":float((cbv>0).mean()),
            "ridge_r2_mean":float(rgv.mean()),
            "catboost_minus_ridge":float(cbv.mean()-rgv.mean()),
            "loyo_r2_median":float(ly.median()),
            "loyo_positive_share":float((ly>0).mean()),
        }
        row["screen_pass"]=bool(
            row["cv_r2_mean"]>0 and
            row["cv_r2_median"]>0 and
            row["catboost_minus_ridge"]>0 and
            row["loyo_positive_share"]>=2/3
        )
        screen_rows.append(row)
    TARGET_SCREEN=pd.DataFrame(screen_rows)
    SCREENED_TARGETS=TARGET_SCREEN.loc[TARGET_SCREEN.screen_pass,"target"].tolist()

    # ------------------------------------------------------------------
    # V5.6. Этап 2: robustness только для прошедших screening целей.
    # ------------------------------------------------------------------
    def cv_reg_score(cols,yv,seed_shift=0,iters=300):
        vals=[]
        for tr,te in KFold(5,shuffle=True,random_state=SEED+seed_shift).split(X[cols],yv):
            im=SimpleImputer(strategy="median",keep_empty_features=True)
            a=im.fit_transform(X.iloc[tr][cols]); b=im.transform(X.iloc[te][cols])
            m=cb_reg(SEED+seed_shift,iters=iters); m.fit(a,yv[tr])
            vals.append(r2_score(yv[te],m.predict(b)))
        return float(np.mean(vals))

    for ti,target in enumerate(SCREENED_TARGETS,1):
        y=XDF[target].to_numpy(float)
        rng=np.random.default_rng(SEED+5000+ti)

        # Bootstrap регионов с OOB оценкой.
        for r in pbar(range(BOOT_N),desc=f"6B BOOT {ti}/{len(SCREENED_TARGETS)} {target}",unit="rep",leave=False):
            idx=rng.integers(0,len(y),len(y))
            oob=np.setdiff1d(np.arange(len(y)),np.unique(idx))
            if len(oob)<12:
                continue
            im=SimpleImputer(strategy="median",keep_empty_features=True)
            a=im.fit_transform(X.iloc[idx]); b=im.transform(X.iloc[oob])
            m=cb_reg(SEED+7000+r+100*ti,iters=350); m.fit(a,y[idx])
            BOOT_ROWS.append({
                "target":target,"rep":r,"n_oob":len(oob),
                "score":r2_score(y[oob],m.predict(b)),"metric":"r2"
            })

        # Negative control: permutation target, +1 correction применяется ниже.
        for r in pbar(range(NEGATIVE_N_FINAL),desc=f"6C NEG {ti}/{len(SCREENED_TARGETS)} {target}",unit="perm",leave=False):
            yp=rng.permutation(y)
            NEG_ROWS.append({
                "target":target,"rep":r,
                "score":cv_reg_score(feature_cols,yp,seed_shift=10000+1000*ti+r,iters=220),
                "metric":"r2"
            })

    BOOTSTRAP=pd.DataFrame(BOOT_ROWS,columns=["target","rep","n_oob","score","metric"])
    NEGATIVE_CONTROL=pd.DataFrame(NEG_ROWS,columns=["target","rep","score","metric"])

    # Строгий научный gate по каждой цели.
    validation=[]
    for _,sr in TARGET_SCREEN.iterrows():
        target=sr.target
        b=BOOTSTRAP[BOOTSTRAP.target.eq(target)].score.astype(float)
        n=NEGATIVE_CONTROL[NEGATIVE_CONTROL.target.eq(target)].score.astype(float)
        real=float(sr.cv_r2_mean)
        neg_p=float((1+(n>=real).sum())/(1+len(n))) if len(n) else np.nan
        bmed=float(b.median()) if len(b) else np.nan
        bpos=float((b>0).mean()) if len(b) else np.nan
        validated=bool(
            sr.screen_pass and
            pd.notna(bmed) and bmed>0 and
            pd.notna(bpos) and bpos>=.60 and
            pd.notna(neg_p) and neg_p<.05
        )
        validation.append({
            **sr.to_dict(),
            "bootstrap_n":int(len(b)),
            "bootstrap_r2_median":bmed,
            "bootstrap_positive_share":bpos,
            "negative_n":int(len(n)),
            "negative_p":neg_p,
            "validated":validated,
        })
    TARGET_VALIDATION=pd.DataFrame(validation)
    VALIDATED_TARGETS=TARGET_VALIDATION.loc[TARGET_VALIDATION.validated,"target"].tolist()
    NEGATIVE_P_BY_TARGET=TARGET_VALIDATION[["target","negative_p"]].copy()
    pipeline_done("model_validation",f"валидация целей: кандидатов={len(SCREENED_TARGETS)}; прошли={len(VALIDATED_TARGETS)}")

    # ------------------------------------------------------------------
    # V5.6. Этап 3: интерпретация только валидированных целей.
    # ------------------------------------------------------------------
    IMP=SimpleImputer(strategy="median",keep_empty_features=True)
    XI=IMP.fit_transform(X)
    for ti,target in enumerate(VALIDATED_TARGETS,1):
        y=XDF[target].to_numpy(float)

        # Полная линейная модель и SHAP.
        sc=StandardScaler(); XIS=sc.fit_transform(XI)
        rg=Ridge(alpha=1.0).fit(XIS,y)
        LINEAR_ROWS.extend({
            "target":target,"class":"regression","feature":f,"coefficient":float(v)
        } for f,v in zip(feature_cols,rg.coef_))

        cb=cb_reg(SEED+20000+ti,iters=600); cb.fit(XI,y)
        sv=cb.get_feature_importance(Pool(XI,label=y),type="ShapValues")
        sa=np.mean(np.abs(sv[:,:-1]),axis=0)
        SHAP_ROWS.extend({
            "target":target,"feature":f,"mean_abs_shap":float(v)
        } for f,v in zip(feature_cols,sa))

        # Permutation importance в честном 5-fold.
        cv=list(KFold(5,shuffle=True,random_state=SEED+30000+ti).split(X,y))
        for f,(tr,te) in enumerate(pbar(cv,desc=f"6D PERM {target}",unit="fold",leave=False)):
            im=SimpleImputer(strategy="median",keep_empty_features=True)
            a=im.fit_transform(X.iloc[tr]); b=im.transform(X.iloc[te])
            m=cb_reg(SEED+31000+f+100*ti,iters=450); m.fit(a,y[tr])
            base=r2_score(y[te],m.predict(b))
            rng=np.random.default_rng(SEED+32000+f+100*ti)
            for j,c in enumerate(feature_cols):
                ds=[]
                for _ in range(12):
                    bp=b.copy(); bp[:,j]=rng.permutation(bp[:,j])
                    ds.append(base-r2_score(y[te],m.predict(bp)))
                PERM_ROWS.append({
                    "target":target,"fold":f,"feature":c,
                    "importance":float(np.mean(ds)),"sd":float(np.std(ds))
                })

        # Абляция пяти теоретических блоков.
        full=cv_reg_score(feature_cols,y,seed_shift=40000+ti,iters=400)
        ABL_ROWS.append({
            "target":target,"removed_block":"none","available":True,
            "features":len(feature_cols),"score":full,"metric":"r2","delta":0.0
        })
        for bl in PREDICTOR_BLOCKS:
            rem=blocks.get(bl,[])
            use=[c for c in feature_cols if c not in rem]
            scv=cv_reg_score(use,y,seed_shift=41000+100*ti+PREDICTOR_BLOCKS.index(bl),iters=400) if rem else np.nan
            ABL_ROWS.append({
                "target":target,"removed_block":bl,"available":bool(rem),
                "features":len(use),"score":scv,"metric":"r2",
                "delta":full-scv if pd.notna(scv) else np.nan
            })

    PERMUTATION=pd.DataFrame(PERM_ROWS,columns=["target","fold","feature","importance","sd"])
    if len(PERMUTATION):
        PERMUTATION=(PERMUTATION.groupby(["target","feature"])
                     .agg(importance=("importance","mean"),sd=("sd","mean")).reset_index())
    SHAP=pd.DataFrame(SHAP_ROWS,columns=["target","feature","mean_abs_shap"])
    LINEAR_COEF=pd.DataFrame(LINEAR_ROWS,columns=["target","class","feature","coefficient"])
    ABLATION=pd.DataFrame(ABL_ROWS,columns=["target","removed_block","available","features","score","metric","delta"])

    if len(PERMUTATION):
        PERMUTATION=PERMUTATION.merge(FEATURE_META,on="feature",how="left")
    if len(SHAP):
        SHAP=SHAP.merge(FEATURE_META,on="feature",how="left")
    if len(LINEAR_COEF):
        LINEAR_COEF=LINEAR_COEF.merge(FEATURE_META,on="feature",how="left")

    pipeline_done("interpretability",f"валидированных целей={len(VALIDATED_TARGETS)}")

    # Совместимые итоговые показатели без ложного требования «положительной PC1».
    if VALIDATED_TARGETS:
        _best=(TARGET_VALIDATION[TARGET_VALIDATION.validated]
               .sort_values("cv_r2_mean",ascending=False).iloc[0])
    else:
        _best=TARGET_VALIDATION.sort_values("cv_r2_mean",ascending=False).iloc[0]
    PRIMARY_MODEL_TARGET=str(_best.target)
    MODEL_PRIMARY_SCORE=float(_best.cv_r2_mean)
    MODEL_BASELINE_SCORE=float(_best.ridge_r2_mean)
    NEGATIVE_P=float(_best.negative_p) if pd.notna(_best.negative_p) else np.nan

    XDF.to_parquet(PROCESSED/"predictors.parquet",index=False)
    PREDICTIONS.to_csv(PROCESSED/"model_predictions.csv",index=False)
    FEATURE_META.to_csv(PROCESSED/"feature_dictionary.csv",index=False)
    PREDICTOR_SELECTION.to_csv(PROCESSED/"selected_predictors.csv",index=False)
    TARGET_SCREEN.to_csv(PROCESSED/"target_screen.csv",index=False)
    TARGET_VALIDATION.to_csv(PROCESSED/"target_validation.csv",index=False)

    # Публикационные графики — только подтвержденные интерпретационные цели.
    RU_BLOCK_SHORT={"demography":"Демография","income":"Доходы","labour":"Рынок труда","industry":"Экономическая структура","social_support":"Социальная поддержка"}
    for target in VALIDATED_TARGETS:
        safe=re.sub(r"[^A-Za-z0-9_]+","_",target)
        if len(SHAP):
            q=SHAP[SHAP.target.eq(target)].nlargest(18,"mean_abs_shap").sort_values("mean_abs_shap")
            if len(q):
                plt.figure(figsize=(11,7.2)); plt.barh(q.label,q.mean_abs_shap)
                plt.xlabel("Среднее абсолютное значение SHAP"); plt.ylabel(""); plt.tight_layout()
                for ext in ("png","svg"):
                    plt.savefig(FIG/f"shap_{safe}.{ext}",dpi=600 if ext=="png" else None,bbox_inches="tight")
                plt.close()
        if len(ABLATION):
            q=ABLATION[(ABLATION.target.eq(target))&(ABLATION.removed_block!="none")&ABLATION.available].copy()
            q["label"]=q.removed_block.map(RU_BLOCK_SHORT); q=q.sort_values("delta")
            if len(q):
                plt.figure(figsize=(8.5,5.2)); plt.barh(q.label,q.delta); plt.axvline(0,color="black",lw=.8)
                plt.xlabel("Снижение R² после исключения блока"); plt.tight_layout()
                for ext in ("png","svg"):
                    plt.savefig(FIG/f"ablation_{safe}.{ext}",dpi=600 if ext=="png" else None,bbox_inches="tight")
                plt.close()

    timing_end(
        "04_explanatory_models_and_robustness",_STAGE4_T0,
        detail=f"targets={len(TARGETS)}; screened={len(SCREENED_TARGETS)}; validated={len(VALIDATED_TARGETS)}; features={len(feature_cols)}"
    )
    print("\nПЕРВИЧНЫЙ SCREENING ЦЕЛЕЙ:"); display(TARGET_SCREEN)
    print("\nСТРОГАЯ ВАЛИДАЦИЯ ЦЕЛЕЙ:"); display(TARGET_VALIDATION)
    if VALIDATED_TARGETS:
        print("Для SHAP/permutation/абляций допущены:",", ".join(VALIDATED_TARGETS))
    else:
        print("Ни одна цель не прошла строгий gate; факторная интерпретация не выполнялась.")
```

```python
# 6. Таблицы для статьи + компактный пакет воспроизводимости V5.6
if not globals().get("CONTINUE_RESEARCH",False):
    print("Блок пропущен.")
else:
    _STAGE6_T0=timing_start("06_article_and_reproducibility_outputs")

    # Сводки для статьи.
    cv_summary=(CV_METRICS.groupby(["target","model"])
                .agg(r2_mean=("r2","mean"),r2_median=("r2","median"),
                     r2_positive_share=("r2",lambda s:float((s>0).mean())),
                     mae_mean=("mae","mean"),folds=("fold","count"))
                .reset_index())
    loyo_summary=(TEMPORAL.groupby("target")
                  .agg(r2_median=("r2","median"),
                       positive_share=("r2",lambda s:float((s>0).mean())),
                       held_years=("held_year","nunique")).reset_index())
    if len(BOOTSTRAP):
        boot_summary=(BOOTSTRAP.groupby("target")
                      .agg(n=("score","count"),median=("score","median"),mean=("score","mean"),
                           q025=("score",lambda s:float(s.quantile(.025))),
                           q975=("score",lambda s:float(s.quantile(.975))),
                           positive_share=("score",lambda s:float((s>0).mean())))
                      .reset_index())
    else:
        boot_summary=pd.DataFrame(columns=["target","n","median","mean","q025","q975","positive_share"])

    article_xlsx=RESULTS/"ARTICLE_TABLES_V5_6.xlsx"
    with pd.ExcelWriter(article_xlsx,engine="xlsxwriter") as w:
        tables={
            "SCIENTIFIC_SUMMARY":SCIENTIFIC_SUMMARY,
            "TARGET_VALIDATION":TARGET_VALIDATION,
            "OUTCOME_INDICATORS":SELECTED,
            "DIRECTION_AUDIT":DIRECTION_AUDIT,
            "REGION_PCA":PCA_COORD,
            "REGION_BLOCK_AXES":BLOCK_TARGETS,
            "PCA_VARIANCE":PCA_VARIANCE,
            "PCA_LOADINGS":PCA_LOADINGS,
            "BLOCK_PCA_VARIANCE":BLOCK_PCA_VARIANCE,
            "BLOCK_PCA_LOADINGS":BLOCK_PCA_LOADINGS,
            "PREDICTORS":PREDICTOR_SELECTION,
            "PREDICTOR_BLOCK_AUDIT":PREDICTOR_BLOCK_AUDIT,
            "FEATURE_DICTIONARY":FEATURE_META,
            "CV_SUMMARY":cv_summary,
            "LOYO_SUMMARY":loyo_summary,
            "BOOTSTRAP_SUMMARY":boot_summary,
            "NEGATIVE_P":NEGATIVE_P_BY_TARGET,
            "SHAP_VALIDATED":SHAP,
            "PERM_VALIDATED":PERMUTATION,
            "ABLATION_VALIDATED":ABLATION,
            "EXPERIMENT_AUDIT":EXPERIMENT_AUDIT,
            "DATA_GATE":DATA_GATE,
        }
        for n,df in tables.items():
            if isinstance(df,pd.DataFrame):
                df.to_excel(w,sheet_name=n[:31],index=False)

    # Компактные CSV для повторного анализа результатов.
    research_data=RESULTS/"research_data"
    research_data.mkdir(exist_ok=True)
    small_tables={
        "scientific_summary.csv":SCIENTIFIC_SUMMARY,
        "experiment_audit.csv":EXPERIMENT_AUDIT,
        "target_validation.csv":TARGET_VALIDATION,
        "selected_outcomes.csv":SELECTED,
        "selected_predictors.csv":PREDICTOR_SELECTION,
        "feature_dictionary.csv":FEATURE_META,
        "regional_pca.csv":PCA_COORD,
        "regional_block_axes.csv":BLOCK_TARGETS,
        "pca_variance.csv":PCA_VARIANCE,
        "block_pca_variance.csv":BLOCK_PCA_VARIANCE,
        "cv_summary.csv":cv_summary,
        "loyo_summary.csv":loyo_summary,
        "bootstrap_summary.csv":boot_summary,
        "negative_p.csv":NEGATIVE_P_BY_TARGET,
        "shap_validated.csv":SHAP,
        "permutation_validated.csv":PERMUTATION,
        "ablation_validated.csv":ABLATION,
        "source_audit.csv":SOURCE_AUDIT,
        "source_completeness.csv":SOURCE_COMPLETENESS,
        "source_catalog.csv":SOURCE_CATALOG,
        "data_gate.csv":DATA_GATE,
    }
    for name,df in small_tables.items():
        if isinstance(df,pd.DataFrame):
            df.to_csv(research_data/name,index=False)

    # Конфигурация и зависимости — для воспроизводства нового запуска с нуля.
    config={
        "version":"V5.6",
        "seed":SEED,
        "pre_years":PRE_YEARS,
        "post_years":POST_YEARS,
        "shock_year":SHOCK_YEAR,
        "bootstrap_n":BOOT_N,
        "repeated_cv_repeats":REPEATED_CV_REPEATS,
        "negative_control_n":NEGATIVE_N_FINAL,
        "outcome_blocks":OUTCOME_BLOCKS,
        "predictor_blocks":PREDICTOR_BLOCKS,
        "validated_targets":VALIDATED_TARGETS,
        "analysis_mode":ANALYSIS_MODE,
        "typology_accepted":bool(TYPOLOGY_ACCEPTED),
        "continuum_accepted":bool(CONTINUUM_ACCEPTED),
        "research_accepted":bool(RESEARCH_ACCEPTED),
    }
    (RESULTS/"run_config_v5_6.json").write_text(json.dumps(config,ensure_ascii=False,indent=2),encoding="utf-8")
    requirements=[
        "pandas>=2.2","numpy","requests","beautifulsoup4","lxml","openpyxl","xlrd",
        "xlsxwriter","pyarrow","python-calamine","tqdm","scikit-learn","catboost",
        "matplotlib","joblib","scipy"
    ]
    (RESULTS/"requirements_v5_6.txt").write_text("\n".join(requirements)+"\n",encoding="utf-8")

    readme=f"""V5.6 REPRODUCIBILITY PACKAGE

Основной воспроизводимый объект — ноутбук MONOGRAPH_CH3_REGIONAL_RESILIENCE_ML_v5_6_FINAL.ipynb.
Он сам загружает доступные официальные источники Росстата, парсит их и повторяет эксперимент с нуля.

В этот компактный пакет намеренно НЕ включены:
- сырые скачанные архивы Росстата;
- распакованные XLS/XLSX;
- panel_raw/panel_clean/panel_long parquet;
- полный кэш загрузчика.

Причина: эти файлы воспроизводятся самим ноутбуком и делают архив избыточно большим.

Научный результат:
- дискретная типология: {TYPOLOGY_ACCEPTED};
- непрерывная геометрия: {CONTINUUM_ACCEPTED};
- валидированные модельные цели: {', '.join(VALIDATED_TARGETS) if VALIDATED_TARGETS else 'нет'};
- итог эксперимента: {'принят' if RESEARCH_ACCEPTED else 'требует проверки'}.

Для статьи используйте ARTICLE_TABLES_V5_6.xlsx и FIGURES_EDITABLE_V5_6.xlsx.
"""
    (RESULTS/"README_REPRODUCIBILITY_V5_6.txt").write_text(readme,encoding="utf-8")

    # SHA256 только для компактных результатов.
    compact_files=[
        RESULTS/"CHAPTER3_ANALYSIS_V5_6.xlsx",
        RESULTS/"ARTICLE_TABLES_V5_6.xlsx",
        RESULTS/"EXPERIMENT_AUDIT_V5_6.xlsx",
        RESULTS/"FIGURES_EDITABLE_V5_6.xlsx",
        RESULTS/"run_config_v5_6.json",
        RESULTS/"requirements_v5_6.txt",
        RESULTS/"README_REPRODUCIBILITY_V5_6.txt",
    ]+[p for p in research_data.glob("*.csv")]+[p for p in FIG.glob("*") if p.is_file()]
    compact_files=[p for p in compact_files if p.exists()]

    manifest=[]
    for p in compact_files:
        h=hashlib.sha256()
        with open(p,"rb") as f:
            for chunk in iter(lambda:f.read(1024*1024),b""):
                h.update(chunk)
        manifest.append({
            "file":str(p.relative_to(RESULTS)),
            "bytes":p.stat().st_size,
            "sha256":h.hexdigest()
        })
    MANIFEST=pd.DataFrame(manifest)
    MANIFEST.to_csv(RESULTS/"manifest_sha256_v5_6.csv",index=False)

    repro_zip=Path("/content/MONOGRAPH_CH3_V5_6_REPRODUCIBILITY.zip")
    repro_zip.unlink(missing_ok=True)
    with zipfile.ZipFile(repro_zip,"w",zipfile.ZIP_DEFLATED) as z:
        pack=compact_files+[RESULTS/"manifest_sha256_v5_6.csv"]
        for p in pbar(pack,desc="8/8 КОМПАКТНЫЙ REPRO ZIP",unit="файл",leave=True):
            if p.exists():
                z.write(p,arcname=str(p.relative_to(RESULTS)))

    timing_end("06_article_and_reproducibility_outputs",_STAGE6_T0,
               detail=f"zip={repro_zip.stat().st_size/1024**2:.1f} MB; validated={len(VALIDATED_TARGETS)}")
    RUN_TIMINGS=timing_frame()
    RUN_TIMINGS.to_csv(RESULTS/"RUN_TIMINGS_V5_6.csv",index=False)

    pipeline_done("exports",f"итоговые файлы готовы; zip={repro_zip.stat().st_size/1024**2:.1f} MB")
    pipeline_close()

    print("\n=== V5.6 ЗАВЕРШЕН ===")
    display(SCIENTIFIC_SUMMARY)
    display(TARGET_VALIDATION)
    print("Основной анализ:",RESULTS/"CHAPTER3_ANALYSIS_V5_6.xlsx")
    print("Таблицы для статьи:",article_xlsx)
    print("Редактируемые рисунки:",RESULTS/"FIGURES_EDITABLE_V5_6.xlsx")
    print("Научный аудит:",RESULTS/"EXPERIMENT_AUDIT_V5_6.xlsx")
    print("Компактный пакет воспроизводимости:",repro_zip)
```
